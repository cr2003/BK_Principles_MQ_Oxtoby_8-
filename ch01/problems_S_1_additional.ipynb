#### 39.

A rough estimate of the radius of a nucleus is provided
by the formula $r = kA^{1/3}$, where k is approximately
$1.3 × 10^{-13}$ cm and A is the mass number of the nucleus.
Estimate the density of the nucleus of $^{127}I$ (which has a
nuclear mass of $2.1 × 10^{-22}$ g) in grams per cubic centimeter.
Compare with the density of solid iodine, $4.93\, g\, cm^{-3}$.


```python
# Given parameters
k = 1.3e-13 * u.cm
A = 127
mass_nucleus = 2.1e-22 * u.g
rho_solid = 4.93 * (u.g / (u.cm**3))

# Calculations
radius = k * (A ** (1 / 3))
volume = (4 / 3) * np.pi * (radius**3)
rho_nucleus = mass_nucleus / volume
ratio = rho_nucleus / rho_solid

print("=== Problem 39 Results ===")
print(f"Nuclear Radius         : {radius:.3e}")
print(f"Nuclear Volume         : {volume:.3e}")
print(f"Nuclear Density        : {rho_nucleus:.2e}")
print(f"Density Ratio          : {ratio:.2e} x denser than solid I2")
```

#### 40. 

In a neutron star, gravity causes the electrons to combine
with protons to form neutrons. A typical neutron star has a
mass half that of the sun, compressed into a sphere of
radius 20 km. If such a neutron star contains $6.0 × 10^{56}$
neutrons, calculate its density in grams per cubic centimeter.
Compare this with the density inside a $^{232}Th$ nucleus, in
which 142 neutrons and 90 protons occupy a sphere of
radius $9.1 × 10^{-13} cm$. Take the mass of a neutron to be
$1.675 × 10^{-24}$ g and that of a proton to be $1.673 × 10^{-24}$ g.

```python
# --- Given Data ---

num_neutrons_star = 6.0e56
r_star = 20 * u.km.to(u.cm) * u.cm  # Convert km to cm

n_neutrons_th = 142
n_protons_th = 90
r_nucleus = 9.1e-13 * u.cm

m_neutron = 1.675e-24 * u.g
m_proton = 1.673e-24 * u.g

# 1. Neutron Star Density
mass_star = num_neutrons_star * m_neutron
vol_star = (4 / 3) * np.pi * (r_star**3)
rho_star = mass_star / vol_star

# 2. Thorium Nucleus Density
mass_nucleus = (n_neutrons_th * m_neutron) + (n_protons_th * m_proton)
vol_nucleus = (4 / 3) * np.pi * (r_nucleus**3)
rho_nucleus = mass_nucleus / vol_nucleus

# 3. Ratio
ratio = rho_nucleus / rho_star

print("=== Problem 40 Results ===")
print(f"Neutron Star Density   : {rho_star:.2e}")
print(f"Thorium Nucleus Density: {rho_nucleus:.2e}")
print(f"Ratio (Nucleus / Star) : {ratio:.2f}")
print(f"The Th nucleus is approximately {ratio:.0f} x denser than a neutron star")
```

#### 42.

page 34d

Naturally occurring rubidium (Rb) consists of two isotopes:
$^{85}Rb$ (atomic mass 84.9117) and $^{87}Rb$ (atomic mass
86.9092). The atomic mass of the isotope mixture found
in nature is 85.4678. Calculate the percentage abundances
of the two isotopes in rubidium.

```python
# Define fractional abundance variable x
x = sp.Symbol("x", real=True)

# Given masses
mass_85 = 84.9117
mass_87 = 86.9092
avg_mass = 85.4678

# Set up equation: avg_mass = x * mass_85 + (1 - x) * mass_87
eq = sp.Eq(avg_mass, x * mass_85 + (1 - x) * mass_87)

# Solve for x
fractional_85 = float(sp.solve(eq, x)[0])
fractional_87 = 1 - fractional_85

pct_85 = fractional_85 * 100
pct_87 = fractional_87 * 100

print("=== Problem 42 Results ===")
print(f"Abundance of 85Rb : {pct_85:.2f}%")
print(f"Abundance of 87Rb : {pct_87:.2f}%")
```

A sample of a gaseous binary compound of boron and
chlorine weighing 2.842 g occupies 0.153 L. This sample
is decomposed to give 0.664 g solid boron and enough
gaseous chlorine ($Cl_2$) to occupy 0.688 L at the same temperature
and pressure. Determine the number of atoms of
boron and the number of atoms of chlorine in a molecule
of this compound.

```python
# Atomic masses
M_B = 10.811 * (u.g / u.mol)
M_Cl = 35.453 * (u.g / u.mol)

# Given data
m_compound = 2.842 * u.g
V_compound = 0.153 * u.L
m_B = 0.664 * u.g
V_Cl2 = 0.688 * u.L

# Mass balance
m_Cl = m_compound - m_B

# Moles of elements
n_B = m_B / M_B
n_Cl = m_Cl / M_Cl
n_Cl2 = n_Cl / 2.0

# Using Avogadro's Law: n_compound / V_compound = n_Cl2 / V_Cl2
n_compound = n_Cl2 * (V_compound / V_Cl2)
molar_mass = m_compound / n_compound

# Empirical mass
M_empirical = M_B + M_Cl
multiplier = round((molar_mass / M_empirical).value)

print("=== Problem 43 Results ===")
print(f"Empirical Formula      : BCl")
print(f"Calculated Molar Mass  : {molar_mass.value:.2f} g/mol")
print(f"Molecular Multiplier n : {multiplier}")
print(f"Molecular Formula      : B{multiplier}Cl{multiplier}")
print(f"Atoms of Boron (B)     : {multiplier}")
print(f"Atoms of Chlorine (Cl) : {multiplier}")
```
