#### 15.

page 34b

In J. J. Thompson’s experiment depicted in Figures 1.10
and 1.11, assume that the electrons are initially traveling
at a speed of $8.00 × 10^6 m s^{-1}$ in the +x direction and the
electrostatic plates have a length of $l = 0.10 m$, between
which the electrons are deflected toward the -y direction
(downward) by an electric field of magnitude $200 V m^{-1}$.
(Note that $1 volt = 1 V =  1 J C^{-1}$.) Calculate S, the magnitude
of the downward displacement, if the fluorescent
screen is located 0.90 m beyond the exit of the region
between the two electrostatic plates. No magnetic field is
applied to counteract the electrostatic force. (Caution: L is
not 1.0 m nor 0.9 m; see Figure 1.11.)

```python
# --- Given values from Problem 15 ---
v_x = 8.00e6 * (u.m / u.s)  # Initial horizontal velocity
l = 0.10 * u.m  # Plate length
d_exit = 0.90 * u.m  # Distance beyond exit to screen
E = 200 * (u.V / u.m)  # Electric field magnitude

# --- Physical Constants (astropy) ---
e = const.e.si  # Elementary charge (Coulombs)
m_e = const.m_e  # Electron mass (kg)

# --- Calculations ---
# 1. Effective distance L from the midpoint of the plates (l/2) to the screen
L = (l / 2) + d_exit

# 2. Time spent inside the electrostatic plates
t1 = l / v_x

# 3. Downward acceleration between plates (a_y = e * E / m_e)
a_y = (e * E) / m_e

# 4. Vertical deflection inside the plates (D = 0.5 * a_y * t1^2)
D = 0.5 * a_y * (t1**2)

# 5. Total displacement on the screen using magnification ratio (S = D * 2L / l)
S = D * (2 * L / l)

# --- Display Results ---
print("=== Problem 15 Automated Calculation ===")
print(f"Effective Distance (L): {L.to(u.m):.2f}")
print(f"Time in plates (t1)   : {t1.to(u.s):.3e}")
print(f"Acceleration (a_y)    : {a_y.to(u.m / u.s**2):.3e}")
print(f"Deflection (D)        : {D.to(u.mm):.3f}")
print(f"Total Deflection (S)  : {S.to(u.cm):.3f} ({S.to(u.m):.4f})")
```

#### 16.

In the problem 15 above, what is $v_y$, the y-component of
the electron’s velocity, when it has reached the right edge
of the plates? Note that 
$a_y = \frac{\Delta v_y}{\Delta t}$
(be careful with the
sign). How is the ratio $S/L$ related to $v_x$ and $v_y$, if $v_y$ is
calculated at the right edge of the plates? The distances
S and L are shown in Figure 1.11.

```python
# --- Given values from Problem 15 ---
v_x = 8.00e6 * (u.m / u.s)
l = 0.10 * u.m
d_exit = 0.90 * u.m
E = 200 * (u.V / u.m)

# --- Physical Constants ---
e = const.e.si
m_e = const.m_e

# --- Step 1: Time in plates ---
dt = l / v_x

# --- Step 2: Acceleration and v_y ---
a_y = -(e * E) / m_e
v_y = a_y * dt

# --- Step 3: Ratio S/L ---
L = (l / 2) + d_exit
ratio_S_L = abs(v_y) / v_x
S = ratio_S_L * L

print("=== Problem 16 Results ===")
print(f"y-component of velocity (v_y): {v_y.to(u.m / u.s):.3e}")
print(f"Ratio |v_y| / v_x            : {ratio_S_L.value:.5f}")
print(f"Ratio S / L                  : {ratio_S_L.value:.5f}")
print(f"Calculated Deflection (S)    : {S.to(u.cm):.3f}")
```
