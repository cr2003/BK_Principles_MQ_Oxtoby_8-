#### 25.

page 34c

Compute the mass (in grams) of a single iodine atom if the
relative atomic mass of iodine is 126.90447 on the
accepted scale of atomic masses (based on 12 as the relative
atomic mass of 12C).


```python
# Molar mass / relative atomic mass of Iodine
molar_mass_I = 126.90447 * (u.g / u.mol)

# Calculate mass of a single atom using N_A
mass_per_atom = molar_mass_I / const.N_A

# Alternative calculation using atomic mass unit (u.u) directly in grams
mass_from_u = (126.90447 * u.u).to(u.g)

print("=== Problem 25 Results ===")
print(f"Mass of 1 Iodine atom (via N_A) : {mass_per_atom.to(u.g):.7e}")
print(f"Mass of 1 Iodine atom (via u.u) : {mass_from_u:.7e}")
```

#### 26.

page 34c

Determine the mass (in grams) of exactly 100 million
atoms of fluorine if the relative atomic mass of fluorine is
18.998403 on a scale on which exactly 12 is the relative
atomic mass of 12C.


```python
# Given parameters
num_atoms = 1e8 * u.dimensionless_unscaled  # Exactly 100 million
molar_mass_F = 18.998403 * (u.g / u.mol)

# Calculate total mass in grams
total_mass = (num_atoms / const.N_A) * molar_mass_F
total_mass_in_g = total_mass.to(u.g)

# Alternative method using atomic mass units (u.u)
total_mass_alt = (num_atoms * 18.998403 * u.u).to(u.g)

print("=== Problem 26 Results ===")
print(f"Total Mass (via N_A) : {total_mass_in_g:.6e}")
print(f"Total Mass (via u.u) : {total_mass_alt:.6e}")
```

#### 27. 

Compute the relative molecular
masses of the following
compounds on the 12C scale:

**(a)** $P_4O_{10}$

**(b)** $BrCl$

**(c)** $Ca(NO_3)_2$ 

**(d)** $KMnO_4$

**(e)** $(NH_4)_2SO_4$


```python
# Standard atomic masses (in unified atomic mass units, u)
ram = {
    "P": 30.973762 * u.u,
    "O": 15.9994 * u.u,
    "Br": 79.904 * u.u,
    "Cl": 35.453 * u.u,
    "Ca": 40.078 * u.u,
    "N": 14.0067 * u.u,
    "K": 39.0983 * u.u,
    "Mn": 54.938044 * u.u,
    "H": 1.00794 * u.u,
    "S": 32.065 * u.u,
}

# Molecular calculations
compounds = {
    "(a) P4O10": 4 * ram["P"] + 10 * ram["O"],
    "(b) BrCl": 1 * ram["Br"] + 1 * ram["Cl"],
    "(c) Ca(NO3)2": 1 * ram["Ca"] + 2 * ram["N"] + 6 * ram["O"],
    "(d) KMnO4": 1 * ram["K"] + 1 * ram["Mn"] + 4 * ram["O"],
    "(e) (NH4)2SO4": 2 * ram["N"] + 8 * ram["H"] + 1 * ram["S"] + 4 * ram["O"],
}

print("=== Problem 27 Full Results ===")
for label, mass in compounds.items():
    print(f"{label:<15}: {mass.value:10.5f} u  ==>  {mass.value:.2f} u")
```

one each second for the entire span of an 80-year life. Has

#### 30.

A gold atom has a diameter of $2.88 × 10^{-10}$ m. Suppose
the atoms in 1.00 mol of gold atoms are arranged just
touching their neighbors in a single straight line. Determine
the length of the line.

```python
d_atom = 2.88e-10 * u.m
n_moles = 1.00 * u.mol

# Total atoms and length calculation
total_atoms = n_moles * const.N_A
length_m = total_atoms * d_atom
length_km = length_m.to(u.km)
length_au = length_m.to(u.au)

print("=== Problem 30 Results ===")
print(f"Total length (meters)     : {length_m.value:.3e} m")
print(f"Total length (kilometers) : {length_km.value:.3e} km")
print(f"Total length (AU)         : {length_au.value:.2f} AU")

print("\nAstronomical Comparison:")
print("1,160 Astronomical Units (AU),")
print("over 30 times the distance from the Sun to Pluto,")
print("or about 16 light-hours long.")
```

#### 32. 

Arrange the following in order of increasing mass: 
1.06 mol $SF_4$ ; 117 g $CH_4$; $8.7 × 10^{23}$ molecules
of $Cl_2O_7$ ; and
$417 × 10^{23}$ atoms of argon (Ar).

```python
# Molar masses
molar_masses = {
    "SF4": (32.065 + 4 * 18.9984) * (u.g / u.mol),
    "Cl2O7": (2 * 35.453 + 7 * 15.9994) * (u.g / u.mol),
    "Ar": 39.948 * (u.g / u.mol),
}

# Mass calculations
mass_SF4 = (1.06 * u.mol) * molar_masses["SF4"]
mass_CH4 = 117.0 * u.g
mass_Cl2O7 = (8.7e23 / const.N_A) * molar_masses["Cl2O7"]
mass_Ar = (417e23 / const.N_A) * molar_masses["Ar"]

samples = [
    ("1.06 mol SF4", mass_SF4.to(u.g).value),
    ("117 g CH4", mass_CH4.value),
    ("8.7 x 10^23 molecules Cl2O7", mass_Cl2O7.to(u.g).value),
    ("417 x 10^23 atoms Ar", mass_Ar.to(u.g).value),
]

# Sort by mass
sorted_samples = sorted(samples, key=lambda x: x[1])

print("=== Problem 32 Results ===")
for sample, mass in sorted_samples:
    print(f"{sample:<32}: {mass:8.2f} g")
```

#### 34. 
Gold costs $400 per troy ounce, and 1 troy ounce = 
31.1035 g. Determine the cost of $10.0 cm^{3}$ gold if the
density of gold is $ 19.32\, g\, cm^{-3}$ at room conditions.

```python
# 1. Define 'troy_ounce' as a derived unit based on grams
# 1 Troy Ounce = 31.1035 grams
troy_ounce = u.def_unit(
    ["troy_ounce", "ozt"], 31.1035 * u.g, doc="Troy ounce, used for precious metals."
)

# 2. Define 'dollar' as a new custom base unit (irreducible physical dimension)
dollar = u.def_unit(["dollar", "USD", "$"], doc="United States Dollar")

# 3. Enable the custom units so Astropy recognizes them in string parsing and conversions
u.add_enabled_units([troy_ounce, dollar])
```

```python
# --- Given data ---
gold_price = 400 * (dollar / troy_ounce)  # problem data
gold_price_today = 4_456 * (dollar / troy_ounce)  # 30/08/2026
V = 10.0 * u.cm**3
rho = 19.32 * (u.g / u.cm**3)

# calculate the mass
# rho = m / V ==> m = rho * V
m = rho * V

# Convert gold price to Dollars per Gram
price_per_gram = gold_price.to(dollar / u.g)
price_per_gram_today = gold_price_today.to(dollar / u.g)

# Calculate the cost
cost = price_per_gram * m
cost_today = price_per_gram_today * m


print("=== Problem 34 Results ===")
print(f"Mass {m:.2f}")
print(f"Cost (problem): {cost:,.0f}")
print(f"Cost (updated): {cost_today:,.0f}")
```

#### 35. 
Aluminum oxide ($Al_2O_3$) occurs in nature as a mineral
called corundum, which is noted for its hardness and resistance
to attack by acids. Its density is $3.97 \, g \, cm^{-3}$. Calculate
the number of atoms of aluminum in $15.0 \, cm^3$ corundum.

```python
# Atomic masses
M_Al = 26.9815386 * (u.g / u.mol)
M_O = 15.9994 * (u.g / u.mol)
M_Al2O3 = 2 * M_Al + 3 * M_O

# Given values
density = 3.97 * (u.g / (u.cm**3))
volume = 15.0 * (u.cm**3)

# Calculation
mass = volume * density
moles_Al2O3 = mass / M_Al2O3
moles_Al_atoms = 2 * moles_Al2O3
num_Al_atoms = moles_Al_atoms * const.N_A

print("=== Problem 35 Results ===")
print(f"Mass of Al2O3          : {mass.value:.2f} g")
print(f"Moles of Al2O3         : {moles_Al2O3.value:.4f} mol")
print(f"Number of Al atoms     : {num_Al_atoms.value:.2e} atoms")
```

#### 36. 

Calculate the number of atoms of silicon (Si) in $ 415\, cm^3 $
of the colorless gas disilane at 0°C and atmospheric pressure,
where its density is $0.00278\, g\, cm^{-3}$. The molecular
formula of disilane is $Si_2H_6$.


```python
# Atomic masses
M_Si = 28.085 * (u.g / u.mol)
M_H = 1.00794 * (u.g / u.mol)
M_Si2H6 = 2 * M_Si + 6 * M_H

# Given values
density = 0.00278 * (u.g / (u.cm**3))
volume = 415 * (u.cm**3)

# Calculation
mass = volume * density
moles_Si2H6 = mass / M_Si2H6
moles_Si_atoms = 2 * moles_Si2H6
num_Si_atoms = moles_Si_atoms * const.N_A

print("=== Problem 36 Results ===")
print(f"Mass of M_Si2H6          : {mass.value:.2f} g")
print(f"Moles of M_Si2H6         : {moles_Si2H6.value:.4f} mol")
print(f"Number of Si atoms     : {num_Si_atoms.value:.2e} atoms")
```
