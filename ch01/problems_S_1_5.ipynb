The natural abundances and isotopic masses of the element
silicon (Si) relative to 12C 5 12.00000 are

| Isotope | % Abundance | Isotopic Mass |
| :--- | :--- | :--- |
| $^{28}S_i$ | 92.21 | 27.97693 |
| $^{29}S_i$ | 4.70 | 28.97649 |
| $^{30}S_i$ | 3.09 | 29.97376 |

Calculate the atomic mass of naturally occurring silicon.

```python
# Data from Problem 17
isotopes = [
    {"name": "28Si", "abundance": 0.9221, "mass": 27.97693 * u.u},
    {"name": "29Si", "abundance": 0.0470, "mass": 28.97649 * u.u},
    {"name": "30Si", "abundance": 0.0309, "mass": 29.97376 * u.u},
]

# Calculate weighted atomic mass
avg_mass = sum(iso["abundance"] * iso["mass"] for iso in isotopes)

print(f"Calculated Atomic Mass: {avg_mass.value:.4f} u")
print(f"Rounded (Sig Figs)    : {avg_mass.value:.2f} u")
```

#### 19.

page 34b

Only two isotopes of boron (B) occur in nature; their
atomic masses and abundances are given in the following
table. Complete the table by computing the relative atomic
mass of $^{11}B$ to four significant figures, taking the tabulated
relative atomic mass of natural boron as 10.811.

| Isotope | % Abundance | Isotopic Mass |
| :--- | :--- | :--- |
| $^{10}B$ | 19.61 | 10.013 |
| $^{11}B$ | 80.39 | ? |



```python
# --- Numerical Approach (Astropy) ---
ram_boron = 10.811 * u.u  # relaltive atomic mass
f_10B = 0.1961
m_10B = 10.013 * u.u
f_11B = 0.8039

# Calculate mass of 11B
m_11B = (ram_boron - (f_10B * m_10B)) / f_11B

print("=== Problem 19 Results ===")
print(f"Calculated Mass of 11B : {m_11B.value:.5f} u")
print(f"Rounded (4 Sig Figs)   : {m_11B.value:.2f} u")

# --- Symbolic Approach (SymPy) ---
x = sp.Symbol("m_11B", real=True, positive=True)
eq = sp.Eq(10.811, 0.1961 * 10.013 + 0.8039 * x)
solution = sp.solve(eq, x)[0]

print(f"SymPy Exact Solution   : {float(solution):.5f}")
```

#### 20.

page 34b

More than half of all the atoms in naturally occurring
zirconium are $^{90}Zr$. The other four stable isotopes of zirconium
have the following relative atomic masses and
abundances:

| Isotope | % Abundance | Isotopic Mass |
| :--- | :--- | :--- |
| $^{91}Z_r$ | 11.27 | 90.9056 |
| $^{92}Z_r$ | 17.17 | 91.9050 |
| $^{94}Z_r$ | 17.33 | 93.9063 |
| $^{96}Z_r$ | 2.78 | 95.9083 |




Compute the relative atomic mass of $^{90}Zr$ to four significant
digits, using the tabulated relative atomic mass
91.224 for natural zirconium.

```python
# --- Numerical Solution with Astropy ---
ram_zr = 91.224 * u.u

isotopes = [
    {"abundance": 0.1127, "mass": 90.9056 * u.u},  # 91Zr
    {"abundance": 0.1717, "mass": 91.9050 * u.u},  # 92Zr
    {"abundance": 0.1733, "mass": 93.9063 * u.u},  # 94Zr
    {"abundance": 0.0278, "mass": 95.9083 * u.u},  # 96Zr
]

f_90Zr = 1.0 - sum(iso["abundance"] for iso in isotopes)
other_contrib = sum(iso["abundance"] * iso["mass"] for iso in isotopes)

m_90Zr = (ram_zr - other_contrib) / f_90Zr

print("=== Problem 20 Results ===")
print(f"Calculated Abundance of 90Zr : {f_90Zr * 100:.2f}%")
print(f"Calculated Mass of 90Zr      : {m_90Zr.value:.5f} u")
print(f"Rounded (4 Sig Digits)       : {m_90Zr.value:.2f} u")

# --- Symbolic Solution with SymPy ---
m = sp.Symbol("m_90Zr", real=True, positive=True)
eq = sp.Eq(
    91.224,
    (0.5145 * m)
    + (0.1127 * 90.9056)
    + (0.1717 * 91.9050)
    + (0.1733 * 93.9063)
    + (0.0278 * 95.9083),
)
solution = sp.solve(eq, m)[0]

print(f"SymPy Exact Solution        : {float(solution):.5f}")
```

#### 21.

The isotope of plutonium used for nuclear fission is $^{239}Pu$.
Determine 

**(a)** the ratio of the number of neutrons in a
$^{239}Pu$ nucleus to the number of protons, and 

**(b)** the number
of electrons in a single plutonium atom.

```python
# --- Atomic properties for 239Pu ---
Z = 94  # Atomic number (protons)
A = 239  # Mass number

# Neutrons and Electrons
N = A - Z  # Number of neutrons
num_electrons = Z  # In a neutral atom

# Ratio of neutrons to protons
ratio = N / Z

print("=== Problem 21 Results ===")
print(f"Number of Protons (Z) : {Z}")
print(f"Number of Neutrons (N): {N}")
print(f"(a) Ratio N / Z       : {ratio:.4f} (Exact: {N}/{Z})")
print(f"(b) Number of Electrons: {num_electrons}")
```
