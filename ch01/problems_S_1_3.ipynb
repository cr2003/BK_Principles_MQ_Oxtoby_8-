page 34a

Nitrogen (N) and silicon (Si) form two binary compounds
with the following compositions:

| Compound | Mass % N | Mass % Si |
| :--- | :--- | :--- |
| 1 | 33.28 | 66.72 |
| 2 | 39.94 | 60.06 |

**(a)** Compute the mass of silicon that combines with
1.0000 g of nitrogen in each case.

**(b)** Show that these compounds satisfy the law of multiple
proportions. If the second compound has the formula
$Si_3N_4$, what is the formula of the first compound?

```python
# --- Given Data from Problem 7 ---
# Compound 1
mass_pct_N_1 = 33.28 * u.percent
mass_pct_Si_1 = 66.72 * u.percent

# Compound 2
mass_pct_N_2 = 39.94 * u.percent
mass_pct_Si_2 = 60.06 * u.percent

# Fixed mass of Nitrogen for part (a)
m_N_fixed = 1.0000 * u.g

# --- Part (a): Mass of Si per 1.0000 g N ---
m_Si_1 = m_N_fixed * (mass_pct_Si_1 / mass_pct_N_1)
m_Si_2 = m_N_fixed * (mass_pct_Si_2 / mass_pct_N_2)

print("=== Part (a) ===")
print(f"Mass of Si in Compound 1: {m_Si_1:.4f}")
print(f"Mass of Si in Compound 2: {m_Si_2:.4f}\n")

# --- Part (b): Law of Multiple Proportions & Formula ---
# 1. Ratio of Si masses per fixed N mass
si_ratio = (m_Si_1 / m_Si_2).decompose()
fractions_ratio = si_ratio.value  # Convert unitless Quantity to float

print("=== Part (b) ===")
print(f"Ratio of Si masses (Compound 1 / Compound 2): {fractions_ratio:.4f}")

# Convert decimal ratio to fractional simple whole numbers (4/3)
# multiplying by 3 gives 4
ratio_factor = fractions_ratio  # ~ 4/3

# Given Compound 2 formula: Si3 N4
si_atoms_comp2 = 3
n_atoms_comp2 = 4

# Determine Si atoms in Compound 1 for the same 4 Nitrogen atoms:
# Extract scalar float value so integer math works correctly
si_atoms_comp1 = float((si_atoms_comp2 * ratio_factor))


print(f"Ratio simplifies to approximately: 4 / 3")
print(f"For 4 N atoms:")
print(f"  Compound 2 has {si_atoms_comp2} Si atoms -> Si3N4")
print(f"  Compound 1 has {si_atoms_comp1:.0f} Si atoms -> Si4N4")

# Simplify empirical formula
empirical_si = int(round(si_atoms_comp1 / n_atoms_comp2))
empirical_n = int(round(n_atoms_comp2 / n_atoms_comp2))

print(f"Empirical Formula for Compound 1: Si{empirical_si}N{empirical_n} (SiN)")
```

page 34a

Vanadium (V) and oxygen (O) form a series of compounds
with the following compositions:

| Compound | Mass % V | Mass % O |
| :--- | :--- | :--- |
| 1 | 76.10 | 23.90 |
| 2 | 67.98 | 32.02 |
| 3 | 61.42 | 38.58 |
| 4 | 56.02 | 43.98 |

What are the relative numbers of atoms of oxygen in the
compounds for a given mass of vanadium?

```python
from fractions import Fraction

# --- Data from Problem 9 ---
compounds = [
    {"pct_V": 76.10 * u.percent, "pct_O": 23.90 * u.percent},
    {"pct_V": 67.98 * u.percent, "pct_O": 32.02 * u.percent},
    {"pct_V": 61.42 * u.percent, "pct_O": 38.58 * u.percent},
    {"pct_V": 56.02 * u.percent, "pct_O": 43.98 * u.percent},
]

# Fixed mass of Vanadium (1.0000 g)
m_V_fixed = 1.0000 * u.g

# Step 1: Calculate mass of O per 1.0000 g V
masses_O = [(m_V_fixed * (c["pct_O"] / c["pct_V"])).value for c in compounds]

# Step 2: Normalize relative to smallest value (Compound 1)
min_m = min(masses_O)
normalized_ratios = [m / min_m for m in masses_O]

# Step 3: Convert floats to simplified Fraction objects
fractions_O = [Fraction(r).limit_denominator(10) for r in normalized_ratios]

print("=== Relative Fractions & Empirical Formulas ===")
for i, (r_float, frac) in enumerate(zip(normalized_ratios, fractions_O), start=1):
    num = frac.numerator
    den = frac.denominator

    # Formula logic:
    # Compound 1 has 1 O per 1 V (V1O1).
    # Compound i has (num / den) * O per 1 V -> V_(den) O_(num)
    v_subscript = den if den > 1 else ""
    o_subscript = num if num > 1 else ""
    formula = f"V{v_subscript}O{o_subscript}"

    print(
        f"Compound {i}: Float = {r_float:.3f} -> Fraction = {frac} -> Formula = {formula}"
    )

    print(f"Smallest Whole-Number Relative Ratio: {(r_float * 2):.0f}")
    print()
```

#### 11

A liquid compound containing only hydrogen and oxygen
is placed in a flask. Two electrodes are dipped into the
liquid and an electric current is passed between them.
Gaseous hydrogen forms at one electrode and gaseous
oxygen at the other. After a time, 14.4 mL hydrogen has
evolved at the negative terminal, and 14.4 mL oxygen has
evolved at the positive terminal.

**(a)** Assign a chemical formula to the compound in the
cell.

**(b)** Explain why more than one formula is possible as the
answer to part (a).


```python
# Given volumes
v_H2 = 14.4 * u.mL
v_O2 = 14.4 * u.mL

# Avogadro's law: Volume ratio = Molar ratio
molar_ratio = (v_H2 / v_O2).decompose()
vol_fraction = Fraction(float(molar_ratio.value)).limit_denominator(10)

print("(a)")
print(f"Volume Ratio (H2 : O2) = {vol_fraction}")

# Each H2 molecule has 2 H atoms; each O2 molecule has 2 O atoms
atom_ratio_H_to_O = vol_fraction  # (2 * H) / (2 * O) = 1 : 1

print(
    f"Atomic Ratio (H : O)   = {atom_ratio_H_to_O.numerator} : {atom_ratio_H_to_O.denominator}"
)

print("\nPossible Formulas (H_n O_n):")
for n in range(1, 4):
    h_sub = n if n > 1 else ""
    o_sub = n if n > 1 else ""
    print(f"  n = {n}: H{h_sub}O{o_sub}")

print("(b)")
print("All would give H2 and O2 in 1:1 ratio.")
```

#### 13.

Pure nitrogen dioxide ($NO_2$) forms when dinitrogen oxide
$(N_2O)$ and oxygen ($O_2$) are mixed in the presence of a
certain
catalyst. What volumes of $N_2O$ and oxygen are
needed to produce 4.0 L $NO_2$ if all gases are held at the
same conditions of temperature and pressure?

```python
# Target volume of NO2 produced
v_NO2 = 4.0 * u.L

# Stoichiometric coefficients from the balanced equation:
# 2 N2O + 3 O2 -> 4 NO2
coeff_N2O = 2
coeff_O2 = 3
coeff_NO2 = 4

# Volume ratios are equal to mole ratios at constant T and P (Gay-Lussac's Law)
v_N2O = v_NO2 * (coeff_N2O / coeff_NO2)
v_O2 = v_NO2 * (coeff_O2 / coeff_NO2)

print("=== Problem 13 Calculations ===")
print(f"Target NO2 Volume : {v_NO2}")
print(f"Required N2O Volume: {v_N2O:.1f}")
print(f"Required O2 Volume : {v_O2:.1f}")
```

#### 14.
Gaseous methanol ($CH_3OH$) reacts with oxygen ($O_2$) to
produce water vapor and carbon dioxide. What volumes
of water vapor and carbon dioxide will be produced from
2.0 L methanol if all gases are held at the same temperature
and pressure conditions?

```python
# First we need to balance de equation: CH3OH + O2 ==>  CO2 + H2O
from sympy import Matrix, lcm

# System of equations for: a CH3OH + b O2 -> c CO2 + d H2O
# [C, H, O] vectors for reactants and products
A = Matrix(
    [
        [1, 0, -1, 0],  # Carbon balance
        [4, 0, 0, -2],  # Hydrogen balance
        [1, 2, -2, -1],  # Oxygen balance
    ]
)

# Find nullspace to get integer coefficients
sol = A.nullspace()[0]
denominators = [val.q for val in sol]
mult = lcm(denominators)
coefficients = [int(val * mult) for val in sol]

print("Coefficients [a, b, c, d]:", coefficients)
# Output: [2, 3, 2, 4]

print("Final balance equation: ")
print(
    f"{coefficients[0]} CH3OH + {coefficients[1]} O2 ==>  {coefficients[2]} CO2 + {coefficients[3]} H2O"
)
```

```python
# Given volume of methanol
v_CH3OH = 2.0 * u.L

# Stoichiometric coefficients from: 2 CH3OH + 3 O2 ->  2 CO2 + 4 H2O
coeff_CH3OH = 2
coeff_CO2 = 2
coeff_H2O = 4

# Calculate volumes of products
v_CO2 = v_CH3OH * (coeff_CO2 / coeff_CH3OH)
v_H2O = v_CH3OH * (coeff_H2O / coeff_CH3OH)

print("=== Problem 14 Calculations ===")
print(f"Methanol Volume      : {v_CH3OH}")
print(f"CO2 Produced         : {v_CO2:.1f}")
print(f"Water Vapor Produced : {v_H2O:.1f}")
```
