One of the heaviest atoms found in nature is $^{238}U$. Its relative
atomic mass is 238.0508 on a scale in which 12 is the atomic
mass of $^{12}C$. Calculate the mass (in grams) of one $^{238}U$ atom.

```python
m = 238.0508 * (u.g / u.mol)

m_one = m / const.N_A
m_one
```


#### Example 1.5
Nitrogen dioxide ($NO_2$) is a major component of urban air pollution. For a sample
containing 4.000 g $NO_2$, calculate 

**(a)** the number of moles of $NO_2$ and 

**(b)** the number of molecules of $NO_2$.


```python
m = 4.000 * u.g
mol_mass_n = 14.007 * (u.g / u.mol)
mol_mass_o = 15.999 * (u.g / u.mol)
mol_mas_no2 = 1 * mol_mass_n + 2 * mol_mass_o

mol_number = m / mol_mas_no2
print(f"(a) number of moles = {mol_number:.5f} of NO_2")

molecules_number = mol_number * const.N_A
print(f"(b) number of molecules = {molecules_number:.4g} of NO_2")
```
