### 2.2.2 Determination of Empirical Formula from Mass Composition
#### Example 2.1

page 37

A 60.00-g sample of a dry-cleaning fluid was analyzed and
found to contain 10.80 g carbon, 1.36 g hydrogen, and 47.84 g
chlorine. Determine the empirical formula of the compound
using a table of atomic masses.

```python
from fractions import Fraction

# --- Given data ---
sample_mass = 60.00 * u.g
C_mass = 10.80 * u.g
H_mass = 1.36 * u.g
Cl_mass = 47.84 * u.g

# --- Atomic masses ---
M_C = 12.011 * (u.g / u.mol)
M_H = 1.008 * (u.g / u.mol)
M_Cl = 35.453 * (u.g / u.mol)

# --- Calculate the amount of each element in the sample ---
carbon = C_mass / M_C
hydrogen = H_mass / M_H
clorine = Cl_mass / M_Cl

# --- Calculate the ratio of elements in the sample ---
ratio_C_Cl = carbon / clorine
fraction_C_Cl = Fraction(float(ratio_C_Cl.value)).limit_denominator(10)
ratio_C_H = carbon / hydrogen
fraction_C_H = Fraction(float(ratio_C_H.value)).limit_denominator(10)

print("=== Example 2.1 Results ===")
print(f"Amount of carbon: {carbon:.4f}")
print(f"Amount of hydrogen: {hydrogen:.4f}")
print(f"Amount of clorine: {clorine:.4f}")

print(f"Ratio carbon / clorine: {fraction_C_Cl}")
print(f"Ratio carbon / hydrogen: {fraction_C_H}")
print(f"The number of moles form the ratio 2:3:3")
print("The empirical formula is: C2H3Cl3")
```

#### Example 2.2

page 38

A certain compound, used as a welding fuel, contains only carbon and hydrogen. Burning
a small sample of this fuel completely in oxygen produces 3.38 g $CO_2$, 0.692 g
water, and no other products. What is the empirical formula of the compound?


```python
from fractions import Fraction

# --- Given data ---
CO2_mass = 3.38 * u.g
H2O_mass = 0.692 * u.g

# --- Atomic masses ---
M_C = 12.011 * (u.g / u.mol)
M_H = 1.008 * (u.g / u.mol)
M_O = 15.999 * (u.g / u.mol)


# --- Calculate the mol of compounds
M_CO2 = 1 * M_C + 2 * M_O
M_H2O = 2 * M_H + 1 * M_O


# --- Calculate the amount of each element in the sample ---
CO2 = CO2_mass / M_CO2
C = CO2  # as the number of C is one to one
H2O = H2O_mass / M_H2O
H = 2 * H2O  # as the number of H is one to two

print("=== Example 2.2 Results ===")
print(f"Amount of carbon: {C:.4f}")
print(f"Amount of hydogen: {H:.4f}")
print(f"The compound contains equal numbers of moles of carbon and hydrogen,")
print(f"we have determined that its empirical formula is CH")
```

### 2.2.4 Connection between Empirical and Molecular Formulas

The density of the welding gas from Example
2.2 is $1.06 \, g \, L^{-1}$ at 25°C and atmospheric pressure. Under the same conditions,
the density of gaseous oxygen (which exists as diatomic $O_2$ molecules with
molar mass of  $32.0 \, g \, mol^{-1}$ is  $1.31 \, g \, L^{-1}$. The approximate molar mass of the
welding gas is, therefore,

```python
# --- Given data ---
rho_welding = 1.06 * (u.g / u.L)  # at 25°C and atmospheric pressure
rho_oxygen = 1.31 * (u.g / u.L)  # at 25°C and atmospheric pressure

# --- Atomic masses ---
M_O2 = 32.0 * (u.g / u.mol)

# --- calculate molar mass of welding gass
M_welding = (rho_welding / rho_oxygen) * M_O2

print("=== Results ===")
print(f"The molar mass of welding gas = {M_welding:.1f}")
print("The molar mass corresponding to the empirical formula CH is 13.0 g / mol")
print("Because 25.9 g / mol is approximately twice this value,")
print("there must be two CH units per molecule;")
print("therefore, the molecular formula is C2H2. The gas is acetylene.")
```

Calcium hypochlorite,$ Ca(OCl)_2$, is used as a bleaching agent. It is produced from
sodium hydroxide, calcium hydroxide, and chlorine according to the following overall
equation:

$$ 2 \, NaOH + \, Ca(OH)_2 \, + \, 2\, Cl_2 \implies Ca(OCl)_2 \, + 2\, NaCl \, + \, 2 \, H_2O $$

How many grams of chlorine and sodium hydroxide react with 1067 g $Ca(OH)_2$, and
how many grams of calcium hypochlorite are produced?

```python
# --- Given data ---
CA_OH_2_mass = 1067 * (u.g)

# --- Atomic masses ---
M_Ca = 40.078 * (u.g / u.mol)
M_O = 15.999 * (u.g / u.mol)
M_H = 1.008 * (u.g / u.mol)
M_Cl = 35.453 * (u.g / u.mol)
M_Na = 22.990 * (u.g / u.mol)


# --- calculate the molar masses of substances
M_Ca_OH_2 = M_Ca + 2 * M_O + 2 * M_H
M_NaOH = M_Na + M_O + M_H
M_Ca_OCl_2 = M_Ca + 2 * M_O + 2 * M_Cl

# --- calculate the number of mols in the reaction
mol_Ca_OH_2 = CA_OH_2_mass / M_Ca_OH_2


# From balanced equation, 1 mol Ca_OH_2 reacts with 2 mol NaOH and 2 mol Cl2 to produce 1 mol Ca_OCl_2
mol_NaOH = mol_Ca_OH_2 * (2 * u.mol / (1 * u.mol))
mol_Cl2 = mol_Ca_OH_2 * (2 * u.mol / (1 * u.mol))
mol_Ca_OCl_2 = mol_Ca_OH_2 * (1 * u.mol / (1 * u.mol))

# --- calculate the masses of the reaction
NaOH_mass = mol_NaOH * M_NaOH
Cl2_mass = mol_Cl2 * (2 * M_Cl)
Ca_OCl_2_mass = mol_Ca_OCl_2 * (M_Ca_OCl_2)


print("=== Results ===")
print(f"mol NaOH = {mol_NaOH:.2f}")
print(f"mol Cl2 = {mol_Cl2:.2f}")
print(f"mol Ca_OCl_2 = {mol_Ca_OCl_2:.2f}")
print()
print(f"Mass NaOH reacting = {NaOH_mass:.0f}")
print(f"Mass Cl2 reacting = {Cl2_mass:.0f}")
print(f"Mass Ca_OCl_2 produced = {Ca_OCl_2_mass:.0f}")
```

#### Example 2.5

Sulfuric acid ($H_2SO_4$) forms in the chemical reaction

$$ 2\, SO_2 \, + \, O_2 \, + \, 2 \, H_2O \implies   2 \, H_2SO_4  $$

Suppose 400 g $SO_2$, 175 g $O_2$, and 125 g $H_2O$ are mixed and the reaction proceeds
until one of the reactants is used up. Which is the limiting reactant? What mass of
$H_2SO_4$ is produced, and what masses of the other reactants remain?

```python
# --- Given data ---
SO_2_mass = 400 * (u.g)
O_2_mass = 175 * (u.g)
H_2O_mass = 125 * (u.g)

# --- Atomic masses using mendeleev library ---
M_S = element("S").atomic_weight * (u.g / u.mol)
M_O = element("O").atomic_weight * (u.g / u.mol)
M_H = element("H").atomic_weight * (u.g / u.mol)

# --- calculate the molar masses of substances
M_SO_2 = M_S + 2 * M_O
M_O_2 = 2 * M_O
M_H_2O = 2 * M_H + M_O
M_H_2SO_4 = 2 * M_H + M_S + 4 * M_O

# --- calculate the number of moles of each reactant
mol_SO_2 = SO_2_mass / M_SO_2
mol_O_2 = O_2_mass / M_O_2
mol_H_2O = H_2O_mass / M_H_2O

# --- calculate the theoretical number of moles of product by each reactant
mol_M_H_2SO_4_by_SO_2 = mol_SO_2 * (
    2 * u.mol / (2 * u.mol)
)  # (2 mol H_2SO_4 / 2 mol SO_2)
mol_M_H_2SO_4_by_O_2 = mol_O_2 * (
    2 * u.mol / (1 * u.mol)
)  # (2 mol H_2SO_4 / 1 mol O_2)
mol_M_H_2SO_4_by_H_2O = mol_H_2O * (
    2 * u.mol / (2 * u.mol)
)  # (2 mol H_2SO_4 / 2 mol H_2O)
# The limiting reactant is SO_2 (smallest amount of final resultant product)

# --- Calculate the moles of reactants and products after the reaction
mol_O_2_remaining = mol_O_2 - (mol_SO_2 * (1 * u.mol / (2 * u.mol)))
mol_H_2O_remaining = mol_H_2O - (mol_SO_2 * (2 * u.mol / (2 * u.mol)))

# --- Calculate the masses of reactants and products after the reaction
mass_H_2SO_4_produced = mol_SO_2 * M_H_2SO_4
mass_O_2_remaining = mol_O_2_remaining * M_O_2
mass_H_2O_remaining = mol_H_2O_remaining * M_H_2O


print("=== Results ===")
print(f"moles of SO_2 = {mol_SO_2:.2f}")
print(f"moles of O_2 = {mol_O_2:.2f}")
print(f"moles of H_2O = {mol_H_2O:.2f}")
print()
print(f"If all the SO_2 reacted, it would give {mol_M_H_2SO_4_by_SO_2:.2f} of H_2SO_4")
print(f"If all the O_2 reacted, it would give {mol_M_H_2SO_4_by_O_2:.2f} of H_2SO_4")
print(f"If all the H_2O reacted, it would give {mol_M_H_2SO_4_by_H_2O:.2f} of H_2SO_4")
print()
print("The limiting reactant is SO_2 (smallest amount of final resultant product)")
print(f"Remaining moles of O_2: {mol_O_2_remaining:.2f}")
print(f"Remaining moles of H_2O: {mol_H_2O_remaining:.2f}")
print()
print("The masses of reactants and products after the reaction are:")
print(f"Mass H_2SO_4 produced : {mass_H_2SO_4_produced:.0f}")
print(f"Mass of O_2 remaining: {mass_O_2_remaining:.0f}")
print(f"Mass of H_2O remaining: {mass_H_2O_remaining:.0f}")
print()
print(
    f"Total mass at the end {mass_H_2SO_4_produced:.0f} + {mass_O_2_remaining:.0f} + {
        mass_H_2O_remaining:.0f} = {
        mass_H_2SO_4_produced + mass_O_2_remaining + mass_H_2O_remaining:.0f}"
)
print(
    f"Total mass originally preseent  {SO_2_mass:.0f} + {O_2_mass:.0f} + {
        H_2O_mass:.0f} = {SO_2_mass + O_2_mass + H_2O_mass:.0f}"
)
```

The sulfide ore of zinc ($ZnS$) is reduced to elemental zinc by
“roasting” it (heating it in air) to give $ZnO$, and then heating
the $ZnO$ with carbon monoxide. The two reactions can be
written as

$$ ZnS \, + \, \frac{3}{2}\, O_2 \implies  ZnO \, + \, SO_2  $$

$$ ZnO \, + \, CO \implies  Zn \, + \, CO_2  $$

Suppose $5.32 kg \, ZnS $ is treated in this way and $3.30 \, kg$ pure
$Zn$ is obtained. Calculate the theoretical yield of zinc and its
actual percentage yield.


```python
# --- Data given ---
ZnS_mass = 5_320 * (u.g)
Zn_mass = 3_300 * (u.g)

# --- Atomic masses using mendeleev library ---
M_Zn = element("Zn").atomic_weight * (u.g / u.mol)
M_S = element("S").atomic_weight * (u.g / u.mol)

# --- calculate the molar masses of substances
M_ZnS = M_Zn + M_S

# --- calculate the number of moles of each reactant
mol_ZnS = ZnS_mass / M_ZnS

# --- Calculate the theoretical number of moles of Zn
# First eq: 1 mol ZnS gives 1 mol ZnO
# Second eq: 1 mol ZnO gives 1 mon Zn
# Theoretical mol yield of zinc from ZnS is 1 : 1 ==> mol Zn = mol ZnS
mol_Zn = mol_ZnS

# --- Calculate the mass of Zn
Zn_mass_th = mol_Zn * M_Zn  # theoretical

# --- Calculate the yield of Zn, ratio = real mass / theoretical mass
pct_yield_Zn = (Zn_mass / Zn_mass_th) * (100 * u.percent)


print("=== Results ===")
print(f"moles of ZnS = {mol_ZnS:.1f}")
print(f"ratio 1:1 ==> moles of Zn = {mol_Zn:.1f}")
print(f"theoretical mass of Zn = {Zn_mass_th:.0f}")
print(f"percentage yield of Zn = {pct_yield_Zn:.1f}")
800701
```
