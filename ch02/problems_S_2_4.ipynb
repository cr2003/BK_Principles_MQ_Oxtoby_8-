21. 

For each of the following chemical reactions, calculate the
mass of the underlined reactant that is required to produce
1.000 g of the underlined product.

**(a)** 

$$ \underline{Mg} + 2\, HCl \implies \underline{H_2} + MgCl_2$$


```python
# Molar masses from mendeleev
mw_mg = element("Mg").atomic_weight * (u.g / u.mol)
mw_h = element("H").atomic_weight * (u.g / u.mol)

target_product_mass = 1.000 * u.g

# part (a) Mg + 2 HCl ==> H2 + MgCl2
mw_h2 = 2 * mw_h
moles_h2 = target_product_mass / mw_h2
moles_mg = moles_h2 * (1 / 1)  # ratio 1/1 between moles of Mg and H2
mass_mg = moles_mg * mw_mg

print(f"(a) Mass of Mg required: {mass_mg:.3f}")
```

**(b)** 

$$ 2\, \underline{CuSO_4} + 4\, KI \implies 2\, CuI + \underline{I_2} + 2\, K_2SO_4$$

```python
# Molar masses from mendeleev
mw_cu = element("Cu").atomic_weight * (u.g / u.mol)
mw_s = element("S").atomic_weight * (u.g / u.mol)
mw_o = element("O").atomic_weight * (u.g / u.mol)
mw_i = element("I").atomic_weight * (u.g / u.mol)


target_product_mass = 1.000 * u.g

# part (b) 2 CuSO4 + 4 KI ==> 2 CuI + 2 K2SO4
mw_cuso4 = mw_cu + mw_s + 4 * mw_o
mw_i2 = 2 * mw_i

moles_I2 = target_product_mass / mw_i2
moles_cuso4 = moles_I2 * (2 / 1)  # ratio 2/1 between moles of CuSO4 and I2
mass_cuso4 = moles_cuso4 * mw_cuso4

print(f"(b) Mass of CuSO4 required: {mass_cuso4:.3f}")
```

**(c)** 

$$ \underline{NaBH_4} + 2\, H_20 \implies NaBO_2 + 4\, \underline{H_2} $$

```python
# Molar masses from mendeleev
mw_na = element("Na").atomic_weight * (u.g / u.mol)
mw_b = element("B").atomic_weight * (u.g / u.mol)

target_product_mass = 1.000 * u.g

# part (C) NaBH4 + 2 H2O ==> NaBO2 + 4 H2
mw_nabh4 = mw_na + mw_b + 4 * mw_h

moles_h2 = target_product_mass / mw_h2
moles_nabh4 = moles_h2 * (1 / 4)  # ratio 1/4 between moles of NaBH4 and H2
mass_nabh4 = moles_nabh4 * mw_nabh4

print(f"(b) Mass of NaBH4 required: {mass_nabh4:.3f}")
```

#### 23. 

An 18.6-g sample of $K_2CO_3$ was treated in such a way that
all of its carbon was captured in the compound
$K_2Zn_3[Fe(CN)_6]_2$. Compute the mass (in grams) of this
product.

```python
mass_k2co3 = 18.6 * u.g

# Molar masses from mendeleev
mw_K = element("K").atomic_weight * (u.g / u.mol)
mw_C = element("C").atomic_weight * (u.g / u.mol)
mw_O = element("O").atomic_weight * (u.g / u.mol)
mw_Zn = element("Zn").atomic_weight * (u.g / u.mol)
mw_Fe = element("Fe").atomic_weight * (u.g / u.mol)
mw_N = element("N").atomic_weight * (u.g / u.mol)

# molecular masses
mw_k2co3 = 2 * mw_K + mw_C + 3 * mw_O
mw_product = 2 * mw_K + 3 * mw_Zn + 2 * mw_Fe + 12 * mw_C + 12 * mw_N


# moles calculation
moles_k2co3 = mass_k2co3 / mw_k2co3
moles_c = moles_k2co3 * 1  # 1 C atom per K2CO3

# Product formed (12 C atoms per product formula unit)
moles_product = moles_c / 12
mass_product = moles_product * mw_product

print(f"Molar mass of K2CO3: {mw_k2co3.value:.3f} g/mol")
print(f"Molar mass of K2Zn3[Fe(CN)6]2: {mw_product.value:.3f} g/mol")
print(f"Moles of K2CO3: {moles_k2co3.value:.5f} mol")
print(f"Moles of product: {moles_product.value:.5f} mol")
print(f"\nCalculated Mass of Product: {mass_product.value:.3f} g")
```

#### 24. 

A chemist dissolves 1.406 g pure platinum (Pt) in an
excess of a mixture of hydrochloric and nitric acids and
then, after a series of subsequent steps involving several
other chemicals, isolates a compound of molecular
formula
$Pt_2C_{10}H{18}N_2S_2O_6$. Determine the maximum possible
yield of this compound.

```python
mass_Pt = 1.406 * u.g

# Molar masses from mendeleev
mw_Pt = element("Pt").atomic_weight * (u.g / u.mol)
mw_C = element("C").atomic_weight * (u.g / u.mol)
mw_H = element("H").atomic_weight * (u.g / u.mol)
mw_N = element("N").atomic_weight * (u.g / u.mol)
mw_S = element("S").atomic_weight * (u.g / u.mol)
mw_O = element("O").atomic_weight * (u.g / u.mol)

# molecular masses
mw_product = 2 * mw_Pt + 10 * mw_C + 18 * mw_H + 2 * mw_N + 2 * mw_S + 6 * mw_O

# moles calculation
moles_Pt = mass_Pt / mw_Pt

# Product formed (2 Pt atoms per product formula unit)
moles_product_max = moles_Pt / 2
mass_product_max = moles_product_max * mw_product

print(f"Molar mass of Pt: {mw_Pt:.3f}")
print(f"Molar mass of product: {mw_product:.3f}")
print(f"Moles of Pt: {moles_Pt:.6f}")
print(f"Max moles of product: {moles_product_max:.6f}")
print(f"\nMaximum Possible Yield: {mass_product_max:.3f}")
```

#### 25. 

Disilane ($Si_2H_6$) is a gas that reacts with oxygen to give
silica ($SiO_2$) and water. Calculate the mass of silica that
would form if 25.0 cm3 disilane (with a density of $ 2.78 × 10^{-3} g cm^{-3}$) reacted with excess oxygen.

```python
# Given data
vol_Si2H6 = 25.0 * (u.cm**3)
rho = 2.78e-3 * (u.g / u.cm**3)

# calculate Si2H6_mass
mass_Si2H6 = rho * vol_Si2H6

# Molar masses from mendeleev
mw_Si = element("Si").atomic_weight * (u.g / u.mol)
mw_H = element("H").atomic_weight * (u.g / u.mol)
mw_O = element("O").atomic_weight * (u.g / u.mol)

# molecular masses
mw_Si2H6 = 2 * mw_Si + 6 * mw_H
mw_SiO2 = mw_Si + 2 * mw_O

# moles calculation
mol_Si2H6 = mass_Si2H6 / mw_Si2H6

# Product formed (1 Si atoms per product per 2 reactant units)
mol_SiO2 = mol_Si2H6 * 2
mass_SiO2 = moles_SiO2 * mw_SiO2


print(f"Mass of Si2H6: {mass_Si2H6:.4f}")
print(f"Moles of Si2H6: {mol_Si2H6:.6f}")
print(f"Moles of SiO2: {mol_SiO2:.6f}")
print(f"\nMass of  SiO2: {mass_SiO2:.3f}")
```

#### 31. 

An element X has a dibromide with the empirical formula
$XBr_2$ and a dichloride with the empirical formula $XCl_2$.
The dibromide is completely converted to the dichloride
when it is heated in a stream of chlorine according to the
reaction

$$XBr_2 + Cl_2 \implies XCl_2 + Br_2$$

When 1.500 g $XBr_2$ is treated, 0.890 g $XCl_2$ results.

**(a)** Calculate the atomic mass of the element X.

**(b)** By reference to a list of the atomic masses of the elements,
identify the element X.

```python
# Standard atomic masses
mw_br = element("Br").atomic_weight * (u.g / u.mol)
mw_cl = element("Cl").atomic_weight * (u.g / u.mol)

# Given masses
mass_xbr2 = 1.500 * u.g
mass_xcl2 = 0.890 * u.g

# Algebraic solution:
# mass_xbr2 / (mx + 2*mw_br) = mass_xcl2 / (mx + 2*mw_cl)
# mass_xbr2 * (mx + 2*mw_cl) = mass_xcl2 * (mx + 2*mw_br)
# mx * (mass_xbr2 - mass_xcl2) = mass_xcl2 * 2 * mw_br - mass_xbr2 * 2 * mw_cl

numerator = (mass_xcl2 * 2 * mw_br) - (mass_xbr2 * 2 * mw_cl)
denominator = mass_xbr2 - mass_xcl2

atomic_mass_x = numerator / denominator

print(f"Atomic Mass of Br: {mw_br.value:.3f} g/mol")
print(f"Atomic Mass of Cl: {mw_cl.value:.3f} g/mol")
print(f"\nCalculated Atomic Mass of X: {atomic_mass_x.value:.2f} g/mol")

# Identify element from periodic table
best_match = None
min_diff = float("inf")

for el in [element("Ni"), element("Co"), element("Fe"), element("Cu")]:
    diff = abs(el.atomic_weight - atomic_mass_x.value)
    if diff < min_diff:
        min_diff = diff
        best_match = el

print(f"Identified Element X: {best_match.name} ({best_match.symbol})")
print(
    f"Standard Atomic Mass of {best_match.symbol}: {best_match.atomic_weight:.3f} g/mol"
)
```

#### 33

A mixture consisting of only sodium chloride (NaCl) and
potassium chloride (KCl) weighs 1.0000 g. When the mixture
is dissolved in water and an excess of silver nitrate is
added, all the chloride ions associated with the original
mixture are precipitated as insoluble silver chloride (AgCl).
The mass of the silver chloride is found to be 2.1476 g.
Calculate the mass percentages of sodium chloride and
potassium chloride in the original mixture.

```python
# 1. Define symbolic variables for unknown masses
x, y = sp.symbols("x y", real=True, positive=True)

# 2. Atomic masses from Mendeleev
mw_na = element("Na").atomic_weight
mw_k = element("K").atomic_weight
mw_cl = element("Cl").atomic_weight
mw_ag = element("Ag").atomic_weight

# Compound molar masses
mw_nacl = mw_na + mw_cl
mw_kcl = mw_k + mw_cl
mw_agcl = mw_ag + mw_cl

# Given experiment values
mass_mixture_val = 1.0000
mass_agcl_val = 2.1476

# 3. Formulate equations using SymPy
# Equation 1: Total mass of mixture
eq1 = sp.Eq(x + y, mass_mixture_val)

# Equation 2: Conservation of chloride moles (n_NaCl + n_KCl = n_AgCl)
eq2 = sp.Eq((x / mw_nacl) + (y / mw_kcl), mass_agcl_val / mw_agcl)

# 4. Solve system of linear equations symbolically
sol = sp.solve((eq1, eq2), (x, y))

mass_nacl = float(sol[x])
mass_kcl = float(sol[y])

# 5. Mass percentages
pct_nacl = (mass_nacl / mass_mixture_val) * 100
pct_kcl = (mass_kcl / mass_mixture_val) * 100

print(f"Molar mass of NaCl: {mw_nacl:.3f} g/mol")
print(f"Molar mass of KCl:  {mw_kcl:.3f} g/mol")
print(f"Molar mass of AgCl: {mw_agcl:.3f} g/mol")
print("-" * 35)
print(f"Calculated Mass of NaCl: {mass_nacl:.4f} g")
print(f"Calculated Mass of KCl:  {mass_kcl:.4f} g")
print("-" * 35)
print(f"Mass Percentage of NaCl: {pct_nacl:.2f}%")
print(f"Mass Percentage of KCl:  {pct_kcl:.2f}%")
```

#### 34. 

A mixture of aluminum and iron weighing 9.62 g reacts
with hydrogen chloride in aqueous solution according to
the parallel reactions
$$ 2\, Al + 6\, HCl \implies 2\, AlCl_3 + 3\, H_2$$

$$ Fe + 2\, HCl \implies FeCl_2 + H_2$$

A 0.738-g quantity of hydrogen is evolved when the metals
react completely. Calculate the mass of iron in the
original mixture.

```python
# 1. Define symbolic variables
x, y = sp.symbols("x y", real=True, positive=True)  # x = mass Al, y = mass Fe

# 2. Atomic masses from Mendeleev
mw_al = element("Al").atomic_weight
mw_fe = element("Fe").atomic_weight
mw_h = element("H").atomic_weight
mw_h2 = 2 * mw_h

# Given experiment values
mass_mixture_val = 9.62
mass_h2_val = 0.738

# Moles of H2 evolved
n_h2_val = mass_h2_val / mw_h2

# 3. Formulate system of equations
# Equation 1: Total mixture mass
eq1 = sp.Eq(x + y, mass_mixture_val)

# Equation 2: Total moles of H2 generated
# (3/2) mol H2 per mol Al, (1/1) mol H2 per mol Fe
eq2 = sp.Eq((1.5 * (x / mw_al)) + (1.0 * (y / mw_fe)), n_h2_val)

# 4. Solve symbolically using SymPy
sol = sp.solve((eq1, eq2), (x, y))

mass_al = float(sol[x])
mass_fe = float(sol[y])

print(f"Atomic Mass of Al: {mw_al:.3f} g/mol")
print(f"Atomic Mass of Fe: {mw_fe:.3f} g/mol")
print(f"Molar Mass of H2:  {mw_h2:.3f} g/mol")
print("-" * 35)
print(f"Calculated Mass of Al: {mass_al:.2f} g")
print(f"Calculated Mass of Fe: {mass_fe:.2f} g")
```
