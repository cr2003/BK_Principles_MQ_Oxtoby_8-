When ammonia is mixed with hydrogen chloride ($HCl$),
the white solid ammonium chloride ($NH_4Cl$) is produced.
Suppose 10.0 g ammonia is mixed with the same mass of
hydrogen chloride. What substances will be present
after
the reaction has gone to completion, and what will their
masses be?


```python
mass_NH3 = 10.0 * u.g
mass_HCl = 10.0 * u.g

# Molar masses from mendeleev
mw_N = element("N").atomic_weight * (u.g / u.mol)
mw_H = element("H").atomic_weight * (u.g / u.mol)
mw_Cl = element("Cl").atomic_weight * (u.g / u.mol)

# molecular masses
mw_NH3 = mw_N + 3 * mw_H
mw_HCl = mw_H + mw_Cl
mw_NH4Cl = mw_N + 4 * mw_H + mw_Cl

# moles calculation
moles_NH3 = mass_NH3 / mw_NH3
moles_HCl = mass_HCl / mw_HCl

moles_NH4Cl = moles_HCl  # 1 mol NH3 produces 1 mol NH4Cl (ratio 1:1), limiting reactant

# Product mass calculation
mass_NH4Cl = moles_NH4Cl * mw_NH4Cl

# Reactant left
NH3_mol_left = moles_NH3 - moles_HCl
NH3_mass_left = NH3_mol_left * mw_NH3


print(f"Molar mass of NH3: {mw_NH3:.3f}")
print(f"Molar mass of HCl: {mw_HCl:.3f}")
print(f"Molar mass of NH4Cl: {mw_NH4Cl:.3f}")
print(f"Moles of NH3: {moles_NH3:.5f} ")
print(f"Moles of HCl: {moles_HCl:.5f} ")
print(
    f"\nThe limiting Reactant is HCl, so the number of moles of product NH4Cl: {moles_NH4Cl:.5f} is the same as HCl"
)

print(f"\nCalculated Mass of Product NH4Cl Yield: {mass_NH4Cl:.1f} ")
print(f"Calculated Mass of Reactant NH3 left: {NH3_mass_left:.1f} ")
```

**(b)** Suppose 500.0 g methane is mixed with 200.0 g ammonia.
Calculate the masses of the substances present
after
the reaction is allowed to proceed to completion.


```python
# Initial masses
mass_CH4 = 500.0 * u.g
mass_NH3 = 200.0 * u.g

# Atomic weights from mendeleev
mw_C = element("C").atomic_weight * (u.g / u.mol)
mw_H = element("H").atomic_weight * (u.g / u.mol)
mw_N = element("N").atomic_weight * (u.g / u.mol)

# molecular masses
mw_CH4 = mw_C + 4 * mw_H
mw_NH3 = mw_N + 3 * mw_H
mw_HCN = mw_H + mw_C + mw_N
mw_H2 = 2 * mw_H

# moles calculation
moles_NH3 = mass_NH3 / mw_NH3
moles_CH4 = mass_CH4 / mw_CH4

# Limiting reactant constraint: xi cannot exceed initial moles of CH4 or NH3, (xi is the extend of reaaction)
xi_max = min(moles_NH3, moles_CH4)

# Remaining / formed moles as functions of xi
moles_CH4_final = moles_CH4 - xi_max
moles_NH3_final = moles_NH3 - xi_max
moles_HCN_final = xi_max
moles_H2_final = 3 * xi_max

# Final masses
mass_CH4_final = moles_CH4_final * mw_CH4
mass_NH3_final = moles_NH3_final * mw_NH3
mass_HCN_final = moles_HCN_final * mw_HCN
mass_H2_final = moles_H2_final * mw_H2

# --- Results ---
print(f"Molar mass of CH4: {mw_CH4:.3f}")
print(f"Molar mass of NH3: {mw_NH3:.3f}")
print(f"Molar mass of HCN: {mw_HCN:.3f}")
print(f"Molar mass of H2: {mw_H2:.3f}")

print("-" * 40)
print(f"The limiting Reactant Extend (xi): {xi_max:.5f}")
print("-" * 40)
print(f"Mass CH4 remaining: {mass_CH4_final:.1f} ")
print(f"Mass NH3 remaining: {mass_NH3_final:.1f} ")
print(f"Mass HCN produced: {mass_HCN_final:.1f} ")
print(f"Mass H2 produced: {mass_H2_final:.1f} ")
print(
    f"Total mass check: {mass_CH4_final + mass_NH3_final + mass_HCN_final + mass_H2_final}"
)
```

#### 37. 

The iron oxide $Fe_2O_3$ reacts with carbon monoxide ($CO$)
to give iron and carbon dioxide:

$$ Fe_2O_3 + 3\, CO \implies 2\, Fe + 3\, CO_2$$

The reaction of 433.2 g $Fe_2O_3$ with excess CO yields
254.3 g iron. Calculate the theoretical yield of iron
(assuming complete reaction) and its percentage yield.

```python
# Initial masses
mass_Fe2O3 = 433.2 * u.g
mass_Fe = 254.3 * u.g

# Atomic weights from mendeleev
mw_Fe = element("Fe").atomic_weight * (u.g / u.mol)
mw_O = element("O").atomic_weight * (u.g / u.mol)

# molecular masses
mw_Fe2O3 = 2 * mw_Fe + 3 * mw_O

# moles calculation
moles_Fe2O3 = mass_Fe2O3 / mw_Fe2O3
moles_Fe_theoretical = (
    2 * moles_Fe2O3
)  # 1 mol of Fe2O3 produce 2 mol of Fe, ratio(1 : 2) by the given equation

# Calculate masses
mass_Fe_theoretical = moles_Fe_theoretical * mw_Fe

# Calculate percentage
pct_Fe = ((mass_Fe * 100) / mass_Fe_theoretical) * u.percent

# --- Results ---
print(f"Molar mass of Fe2O3: {mw_Fe2O3:.3f}")
print(f"Molar mass of Fe: {mw_Fe:.3f}")
print(f"Theoretical moles of Fe: {moles_Fe_theoretical:.3f}")
print(f"Theoretical yield mass of Fe: {mass_Fe_theoretical:.1f}")
print(f"Real mass of Fe yield: {mass_Fe:.1f}")
print(f"Percentage of Fe yield: {pct_Fe:.2f}")
```

#### 38. 

Titanium dioxide, $TiO_2$, reacts with carbon and chlorine
to give gaseous $TiCl_4$:

$$ TiO_2 + 2\, C + 2\, Cl_2 \implies TiCl_4 + 2\, CO$$

The reaction of 7.39 kg titanium dioxide with excess C
and $Cl_2$ gives 14.24 kg titanium tetrachloride. Calculate
the theoretical yield of $TiCl_4$ (assuming complete reaction)
and its percentage yield.

```python
# Initial masses
mass_TiO2 = (7.39 * u.kg).to(u.g)
mass_TiCl4 = (14.24 * u.kg).to(u.g)

# Atomic weights from mendeleev
mw_Ti = element("Ti").atomic_weight * (u.g / u.mol)
mw_O = element("O").atomic_weight * (u.g / u.mol)
mw_Cl = element("Cl").atomic_weight * (u.g / u.mol)


# molecular masses
mw_TiO2 = mw_Ti + 2 * mw_O
mw_TiCl4 = mw_Ti + 4 * mw_Cl


# moles calculation
moles_TiO2 = mass_TiO2 / mw_TiO2
moles_TiCl4_theoretical = moles_TiO2  # 1 mol of TiO2 produce 1 mol of TiCl4, ratio(1 : 1) by the given equation

# Calculate masses
mass_TiCl4_theoretical = moles_TiCl4_theoretical * mw_TiCl4

# Calculate percentage
pct_TiCl4 = ((mass_TiCl4 * 100) / mass_TiCl4_theoretical) * u.percent

# --- Results ---
print(f"Molar mass of TiO2: {mw_TiO2:.3f}")
print(f"Molar mass of TiCl4: {mw_TiCl4:.3f}")

print(f"Theoretical moles of TiCl4: {moles_TiCl4_theoretical:.3f}")
print(f"Theoretical yield mass of TiCl4: {mass_TiCl4_theoretical.to(u.kg):.2f}")
print(f"Real mass of TiCl4 yield: {mass_TiCl4.to(u.kg):.2f}")
print(f"Percentage of TiCl4 yield: {pct_TiCl4:.2f}")
```
