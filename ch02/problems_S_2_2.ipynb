#### 1. 

page 49a

A newly synthesized compound has the molecular formula
$ClF_2O_2PtF_6$. Compute, to four significant figures, the mass
percentage of each of the four elements in this compound.

```python
# Define elements and their atom counts in ClF8O2Pt
elements_count = {"Cl": 1, "F": 8, "O": 2, "Pt": 1}

# Fetch atomic masses as astropy Quantities
masses = {el: element(el).atomic_weight * (u.g / u.mol) for el in elements_count}

# Calculate total molar mass
total_molar_mass = sum(masses[el] * count for el, count in elements_count.items())

print("=== Problem 1: Results ===")
print(f"Total Molar Mass: {total_molar_mass:.4f}\n")
print("=== Mass Percentages (4 Sig Figs) ===")

for el, count in elements_count.items():
    element_mass = masses[el] * count
    pct = (element_mass / total_molar_mass) * (100 * u.percent)
    print(f"{el:2s}: {pct:.4g}")
```

#### 3. 

Arrange the following compounds from left to right in
order of increasing percentage by mass of hydrogen: $H_2O$,
$C_{12}H_{26}$, N_4H_6, Li_H.

```python
# List of compounds represented as dicts of element counts
compounds = {
    "H20": {"H": 2, "O": 1},
    "C12H26": {"C": 12, "H": 26},
    "N4H6": {"N": 4, "H": 6},
    "LiH": {"Li": 1, "H": 1},
}

h_mass = element("H").atomic_weight * (u.g / u.mol)

# Calculate %H for each compound
results = {}
for name, formula in compounds.items():
    total_mass = sum(
        element(el).atomic_weight * count for el, count in formula.items()
    ) * (u.g / u.mol)
    h_total = formula.get("H", 0) * h_mass
    pct_h = (h_total / total_mass) * (100 * u.percent)
    results[name] = pct_h

# Sort by %H ascending
sorted_compounds = sorted(results.items(), key=lambda x: x[1])

print("=== Problem 3: Results ===")
print()
print("--- Mass percentage of Hydrogen --- ")
for name, pct in sorted_compounds:
    print(f"{name:8s}: {pct:.2f}")

print("\nFinal Order (Increasing %H):")
print(" < ".join(name for name, _ in sorted_compounds))
```

#### 6. 

A pharmacist prepares an antiulcer medicine by mixing
286 g $Na_2CO_3$ with water, adding 150 g glycine ($C_2H_5NO_2$),
and stirring continuously at 40°C until a firm mass results.
The pharmacist heats the mass gently until all the water has
been driven away. No other chemical changes occur in this
step. Compute the mass percentage of carbon in the resulting
white crystalline medicine.

```python
# Atomic masses
c_mass = element("C").atomic_weight * (u.g / u.mol)
na_mass = element("Na").atomic_weight * (u.g / u.mol)
o_mass = element("O").atomic_weight * (u.g / u.mol)
h_mass = element("H").atomic_weight * (u.g / u.mol)
n_mass = element("N").atomic_weight * (u.g / u.mol)

# Molar masses
molar_mass_na2co3 = 2 * na_mass + c_mass + 3 * o_mass
molar_mass_glycine = 2 * c_mass + 5 * h_mass + n_mass + 2 * o_mass

# Masses in the mixture
m_na2co3 = 286.0 * u.g
m_glycine = 150.0 * u.g
m_total = m_na2co3 + m_glycine

# Carbon contributions
m_c_na2co3 = m_na2co3 * (c_mass / molar_mass_na2co3)
m_c_glycine = m_glycine * ((2 * c_mass) / molar_mass_glycine)
m_c_total = m_c_na2co3 + m_c_glycine

# Mass percentage of C
pct_c = (m_c_total / m_total).to(u.dimensionless_unscaled) * (100 * u.percent)

print("=== Problem 6: Results ===")
print(f"Total dry mass: {m_total:.1f}")
print(f"Carbon from Na2CO3: {m_c_na2co3:.2f}")
print(f"Carbon from Glycine: {m_c_glycine:.2f}")
print(f"Total Carbon mass: {m_c_total:.2f}")
print(
    f"Mass % of Carbon in medicine: {pct_c.value:.4g}% (or {pct_c.value:.3g}% to 3 sig figs)"
)
```

#### 8. 
Bromoform is 94.85% bromine, 0.40% hydrogen, and
4.75% carbon by mass. Determine its empirical formula.

```python
# Sample size = 100 g
sample_mass = 100.0 * u.g

# Mass fractions
pcts = {"C": 4.75 * u.percent, "H": 0.40 * u.percent, "Br": 94.85 * u.percent}

# Calculate mass and moles
moles = {}
for el, pct in pcts.items():
    mass = sample_mass * (pct.value / 100.0)
    atomic_mass = element(el).atomic_weight * (u.g / u.mol)
    moles[el] = (mass / atomic_mass).to(u.mol)

# Divide by smallest mole value
min_moles = min(moles.values())
ratios = {el: (mol / min_moles).value for el, mol in moles.items()}


print("=== Problem 8: Results ===")

print("=== Moles per 100g sample ===")
for el, mol in moles.items():
    print(f"{el:2s}: {mol.value:.4f} mol")

print("\n=== Mole Ratios ===")
for el, ratio in ratios.items():
    print(f"{el:2s}: {ratio:.2f} -> {round(ratio)}")

formula = "".join(f"{el}{round(r)}" if round(r) > 1 else el for el, r in ratios.items())
print(f"\nEmpirical Formula: {formula}")
```

#### 11. 

Barium and nitrogen form two binary compounds containing
90.745% and 93.634% barium, respectively. Determine
the empirical formulas of these two compounds.

```python
import math

# Atomic weights
ba_mw = element("Ba").atomic_weight * (u.g / u.mol)
n_mw = element("N").atomic_weight * (u.g / u.mol)

compounds = {"Compound 1": 90.745, "Compound 2": 93.634}

for name, pct_ba_val in compounds.items():
    pct_ba = pct_ba_val * u.percent
    pct_n = (100.0 - pct_ba_val) * u.percent

    moles_ba = (pct_ba.to_value(u.dimensionless_unscaled) * u.g / ba_mw).to(u.mol).value
    moles_n = (pct_n.to_value(u.dimensionless_unscaled) * u.g / n_mw).to(u.mol).value

    moles = {"Ba": moles_ba, "N": moles_n}
    min_moles = min(moles.values())

    raw_ratios = {el: mol / min_moles for el, mol in moles.items()}
    frac_ratios = {
        el: Fraction(r).limit_denominator(10) for el, r in raw_ratios.items()
    }
    lcm_denom = math.lcm(*(f.denominator for f in frac_ratios.values()))
    counts = {el: int(f * lcm_denom) for el, f in frac_ratios.items()}

    formula = "".join(f"{el}{c if c > 1 else ''}" for el, c in counts.items())

    print(f"=== {name} ===")
    print(f"Ba moles: {moles_ba:.6f}, N moles: {moles_n:.6f}")
    print(f"Raw Ratio (Ba/N): {moles_ba / moles_n:.4f}")
    print(f"Empirical Formula: {formula}\n")
```

#### 13. 

A sample of 1.000 g of a compound containing carbon
and hydrogen reacts with oxygen at elevated temperature
to yield 0.692 g $H_2O$ and 3.381 g $CO_2$.

**(a)** Calculate the masses of C and H in the sample.

**(b)** Does the compound contain any other elements?

**(c)** What are the mass percentages of C and H in the
compound?

**(d)** What is the empirical formula of the compound?

```python
# Input data
sample_mass = 1.000 * u.g
m_co2 = 3.381 * u.g
m_h2o = 0.692 * u.g

# Atomic and Molecular Weights
mw_c = element("C").atomic_weight * (u.g / u.mol)
mw_h = element("H").atomic_weight * (u.g / u.mol)
mw_o = element("O").atomic_weight * (u.g / u.mol)

mw_co2 = mw_c + 2 * mw_o
mw_h2o = 2 * mw_h + mw_o

# (a) Mass calculations
mass_c = m_co2 * (mw_c / mw_co2)
mass_h = m_h2o * (2 * mw_h / mw_h2o)

print("=== (a) Mass of C and H ===")
print(f"Mass of C: {mass_c:.4f} ")
print(f"Mass of H: {mass_h:.4f} ")

# (b) Other elements check
total_mass = mass_c + mass_h
print("\n=== (b) Other elements check ===")
print(f"Total calculated mass: {total_mass:.4f}  (Sample mass: {sample_mass:.3f} )")
print(
    "Contains other elements?: No"
    if math.isclose(total_mass.value, sample_mass.value, abs_tol=0.005)
    else "Yes"
)

# (c) Mass percentages
pct_c = (mass_c / sample_mass) * 100 * u.percent
pct_h = (mass_h / sample_mass) * 100 * u.percent

print("\n=== (c) Mass Percentages ===")
print(f"C %: {pct_c:.2f}")
print(f"H %: {pct_h:.2f}")

# (d) Empirical Formula
moles_c = (mass_c / mw_c).to(u.mol).value
moles_h = (mass_h / mw_h).to(u.mol).value

moles = {"C": moles_c, "H": moles_h}
min_moles = min(moles.values())

raw_ratios = {el: mol / min_moles for el, mol in moles.items()}
frac_ratios = {el: Fraction(r).limit_denominator(10) for el, r in raw_ratios.items()}
lcm_denom = math.lcm(*(f.denominator for f in frac_ratios.values()))
counts = {el: int(f * lcm_denom) for el, f in frac_ratios.items()}

formula = "".join(f"{el}{c if c > 1 else ''}" for el, c in counts.items())

print("\n=== (d) Empirical Formula ===")
print(f"Raw Ratio C:H -> {moles_c / min_moles:.4f} : {moles_h / min_moles:.4f}")
print(f"Empirical Formula: {formula}")
```

#### 15. 

The empirical formula of a gaseous fluorocarbon is $CF_2$.
At a certain temperature and pressure, a 1-L volume holds
8.93 g of this fluorocarbon, whereas under the same conditions,
the 1-L volume holds only 1.70 g gaseous fluorine
(F2). Determine the molecular
formula of this compound.


```python
# Molar masses from mendeleev
mw_f = element("F").atomic_weight * (u.g / u.mol)
mw_c = element("C").atomic_weight * (u.g / u.mol)

mw_f2 = 2 * mw_f
mw_cf2 = mw_c + 2 * mw_f

# Masses given for 1-L volume under same T, P
m_fluorocarbon = 8.93 * u.g
m_f2 = 1.70 * u.g

# Molar mass of fluorocarbon via Avogadro's hypothesis
mw_fluorocarbon = mw_f2 * (m_fluorocarbon / m_f2)

# Stoichiometric multiple n
n_float = (mw_fluorocarbon / mw_cf2).to_value(u.dimensionless_unscaled)
n = round(n_float)

formula = f"C{n}F{n * 2}"

print(f"Molar mass of F2: {mw_f2.value:.4f} g/mol")
print(f"Calculated Molar mass of fluorocarbon: {mw_fluorocarbon.value:.2f} g/mol")
print(f"Empirical mass (CF2): {mw_cf2.value:.4f} g/mol")
print(f"Multiple n: {n_float:.4f} -> {n}")
print(f"\nMolecular Formula: {formula}")
```

#### 16. 

At its boiling point (280°C) and at atmospheric pressure,
phosphorus has a gas density of $2.7 g L^{-1}$. Under the same
conditions, nitrogen has a gas density of $0.62 g L^{-1}$. How
many atoms of phosphorus are there in one phosphorus
molecule under these conditions?

```python
# Densities
rho_p = 2.7 * (u.g / u.L)
rho_n2 = 0.62 * (u.g / u.L)

# Atomic weights from mendeleev
mw_n = element("N").atomic_weight * (u.g / u.mol)
mw_p_atom = element("P").atomic_weight * (u.g / u.mol)

# Molar mass of nitrogen gas (N2)
mw_n2 = 2 * mw_n

# Molar mass of gaseous phosphorus
mw_p_gas = mw_n2 * (rho_p / rho_n2)

# Number of phosphorus atoms per molecule
atoms_per_molecule_float = (mw_p_gas / mw_p_atom).to_value(u.dimensionless_unscaled)
atoms_per_molecule = round(atoms_per_molecule_float)

print(f"Molar mass of N2: {mw_n2.value:.4f} g/mol")
print(f"Calculated Molar mass of Phosphorus gas: {mw_p_gas.value:.2f} g/mol")
print(f"Atomic weight of P: {mw_p_atom.value:.4f} g/mol")
print(f"Atoms per molecule: {atoms_per_molecule_float:.4f} -> {atoms_per_molecule}")
print(f"\nMolecular Formula: P{atoms_per_molecule}")
```

#### 17. 

A gaseous binary compound has a vapor density that is
1.94 times that of oxygen at the same temperature and
pressure. When 1.39 g of the gas is burned in an excess of
oxygen, 1.21 g water is formed, removing all the hydrogen
originally present.

**(a)** Estimate the molecular mass of the gaseous compound.

```python
# Data given
density_ratio = 1.94
m_compound = 1.39 * u.g
m_h2o = 1.21 * u.g

# Atomic weights
mw_o = element("O").atomic_weight * (u.g / u.mol)
mw_h = element("H").atomic_weight * (u.g / u.mol)

mw_o2 = 2 * mw_o
mw_h2o = 2 * mw_h + mw_o

# (a) Estimated Molecular Mass
mw_compound_est = density_ratio * mw_o2
print("=== (a) Estimated Molecular Mass ===")
print(f"Molar mass of compound: {mw_compound_est.value:.2f} g/mol")
```

**(b)** How many hydrogen atoms are there in a molecule
of the compound?

```python
# (b) Hydrogen count
mass_h = m_h2o * (2 * mw_h / mw_h2o)
moles_h = (mass_h / mw_h).to_value(u.mol)
moles_compound = (m_compound / mw_compound_est).to_value(u.mol)

h_atoms = round(moles_h / moles_compound)
print("\n=== (b) Hydrogen Count ===")
print(f"H atoms per molecule: {h_atoms}")
```

**(c)** What is the maximum possible value of the atomic
mass of the second element in the compound?

```python
# (c) Maximum atomic mass (a = 1)
mass_x_total = mw_compound_est.value - (h_atoms * mw_h.value)
max_atomic_mass_x = mass_x_total / 1
print("\n=== (c) Maximum Atomic Mass ===")
print(f"Max atomic mass of X (a=1): {max_atomic_mass_x:.2f} g/mol")
```

```python
# (d) Candidate elements for different values of 'a'
print("\n=== (d) Possible Atomic Masses for 'a' atoms of X ===")
for a in range(1, 5):
    calc_mw = mass_x_total / a
    print(f"a = {a}: Atomic mass = {calc_mw:.2f} g/mol")

# Theoretical match
mw_si = element("Si").atomic_weight * (u.g / u.mol)
mw_si2h6 = 2 * mw_si + 6 * mw_h

print("\n=== (e) Final Molecular Formula ===")
print(f"Formula: Si2H6 (Disilane)")
print(f"Exact theoretical Molar Mass: {mw_si2h6.value:.2f} g/mol")
```
