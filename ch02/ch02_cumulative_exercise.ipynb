Metallic titanium and its alloys (especially those with aluminum and vanadium)
combine the advantages of high strength and lightweight and are therefore used
widely in the aerospace industry for the bodies and engines of airplanes (Figure
2.4). The major natural source for titanium is the ore rutile, which contains
titanium dioxide ($TiO2$).


**(a)** An intermediate in the preparation of elemental titanium from $TiO2$ is a
volatile chloride of titanium (boiling point 136°C) that contains 25.24%
titanium by mass. Determine the empirical formula of this compound.

```python
# 1. Fetch atomic masses as astropy Quantities
m_Ti = element("Ti").atomic_weight * (u.g / u.mol)
m_Cl = element("Cl").atomic_weight * (u.g / u.mol)

# 2. Define given mass percentages
pct_Ti = 25.24 * u.percent
pct_Cl = (100 * u.percent) - pct_Ti  # 74.76 %

# 3. Assume a 100 g total sample mass
sample_mass = 100 * u.g

mass_Ti = sample_mass * pct_Ti.to(u.dimensionless_unscaled)
mass_Cl = sample_mass * pct_Cl.to(u.dimensionless_unscaled)

# 4. Calculate moles
moles_Ti = (mass_Ti / m_Ti).to(u.mol)
moles_Cl = (mass_Cl / m_Cl).to(u.mol)

# 5. Simplify ratio
ratio_Cl = round((moles_Cl / moles_Ti).value)

print("=== Part (a) Results ===")
print(f"ratio Cl / Ti: {ratio_Cl}")
print(f"Empirical Formula: TiCl{ratio_Cl}")
# Output: Empirical Formula: TiCl4
```

**(b)** At 136°C and atmospheric pressure, the density of this gaseous chloride is
$5.6 g L^{-1}$. Under the same conditions, the density of gaseous nitrogen (N2,
molar mass $28.0 g mol^{-1}$) is $0.83 g L^{-1}$. Determine the molecular
formula of this compound.

```python
# --- Given data ---
rho_compound = 5.6 * (u.g / u.L)
rho_N2 = 0.83 * (u.g / u.L)
M_N2 = 28.0 * (u.g / u.mol)

# --- Calculate Molar Mass of Compound ---
M_compound = (rho_compound / rho_N2) * M_N2

# --- Empirical Formula Mass (TiCl4) ---
M_Ti = element("Ti").atomic_weight * (u.g / u.mol)
M_Cl = element("Cl").atomic_weight * (u.g / u.mol)
M_empirical = M_Ti + (4 * M_Cl)

# --- Determine Ratio n ---
n = round((M_compound / M_empirical).value)

print("=== Part (b) Results ===")
print(f"Molar mass of compound      : {M_compound:.1f}")
print(f"Molar mass of TiCl4 unit    : {M_empirical:.1f}")
print(f"Ratio (n)                   : {n}")
print(f"The molecular formula is    : TiCl4")
```

**(f)** Suppose the reaction chamber for part (e) contains 351 g of the titanium
chloride and 63.2 g liquid magnesium. Which is the limiting reactant? What
maximum mass of titanium could result?

```python
# --- Given data ---
mass_TiCl4 = 351 * (u.g)
mass_Mg = 63.2 * (u.g)


# --- Atomic masses using mendeleev library ---
M_Mg = element("Mg").atomic_weight * (u.g / u.mol)

# --- Moles available ---
n_TiCl4 = mass_TiCl4 / M_TiCl4
n_Mg = mass_Mg / M_Mg

# --- Determine maximum reaction extent (moles of Ti) ---
# Reaction: TiCl4 + 2 Mg -> Ti + 2 MgCl2
max_Ti_from_TiCl4 = n_TiCl4  # 1:1 ratio
max_Ti_from_Mg = n_Mg / 2  # 1:2 ratio

if max_Ti_from_Mg < max_Ti_from_TiCl4:
    limiting = "Mg (Magnesium)"
    n_Ti_max = max_Ti_from_Mg
else:
    limiting = "TiCl4 (Titanium Chloride)"
    n_Ti_max = max_Ti_from_TiCl4

m_Ti_max = (n_Ti_max * M_Ti).to(u.g)

print("=== Part (f) Results ===")
print(f"Moles TiCl4 available : {n_TiCl4:.4f}")
print(f"Moles Mg available    : {n_Mg:.4f}")
print(f"Limiting Reactant     : {limiting}")
print(f"Max yield of Ti       : {m_Ti_max:.1f}")
```
