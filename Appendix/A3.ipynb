Round off each of the measurements
in problem 12 to two
significant figures.

#### 16.

Round off the measured
number in problem 15 to eight,
seven, six, five, four, three, two, and one significant digits.

```python
val = 2997215.548 * u.dimensionless_unscaled

# Iteratively round to 8 down to 1 significant digits using format specifiers
for sf in range(8, 0, -1):
    decimals = sf - 1
    formatted = f"{val.value:.{decimals}e}"
    print(f"{sf} Significant Digits: {formatted}")
```

#### 22

An inch is defined as exactly 2.54 cm. The length of a table
is measured as 505.16 cm. Compute the length of the table
in inches. Justify

```python
l = 505.16 * u.cm
l_inch = l.to(u.imperial.inch)
print(f"length in inches l = {l_inch:.2f} ")
print(f"The measured value ({l}) contains 5 significant figures.")
print(f"Therefore, the result is rounded to 5 significant figures ({l_inch:.2f}).")
```
