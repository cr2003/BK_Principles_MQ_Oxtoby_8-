Calculate the derivatives of the following functions.

**(a)** $y = 4x^2 + 4$

```python
x = sp.Symbol("x")
y = 4 * x**2 + 4

# Symbolic differentiation
dy_dx = sp.diff(y, x)


display(Math(rf"\frac{{dy}}{{dx}} = \mathbf{{{sp.latex(dy_dx)}}}"))
```

**(b)** $y = \sin 3x + 4 \cos 2x$

```python
x = sp.Symbol("x")
y = sp.sin(3 * x) + 4 * sp.cos(2 * x)

# Symbolic differentiation
dy_dx = sp.diff(y, x)

display(Math(rf"\frac{{dy}}{{dx}} = \mathbf{{{sp.latex(dy_dx)}}}"))
```
