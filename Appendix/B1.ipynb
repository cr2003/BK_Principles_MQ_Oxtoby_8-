### B.1 SI Units and Unit Conversions

```python
V = 1 * u.L
print(f"{V} = {V.to(u.dm**3)} =  {V.to(u.m**3):.0e} = {V.to(u.cm**3):.0e}")
```

```python
l = 1 * u.AA
print(f"{l} = {l.to(u.m):.0e} =  {l.to(u.pm):.0f} =  {l.to(u.nm):.1f}")
```

```python
P = 1 * const.atm
print(f"1 atm = {P.to(u.Pa):,.0f} = {P.to(u.MPa):,.6f}  = {P.to(u.hPa):,.2f}")
```

```python
temp_celsius = 0 * u.deg_C
temp_kelvin = temp_celsius.to(u.K, equivalencies=u.temperature())
print(f"{temp_celsius} = {temp_kelvin}")

temp_fahrenheit = temp_celsius.to(u.imperial.Fahrenheit, equivalencies=u.temperature())
print(f"{temp_celsius} = {temp_fahrenheit}")

temp_fahrenheit = 100 * u.imperial.Fahrenheit
temp_celsius = temp_fahrenheit.to(u.deg_C, equivalencies=u.temperature())
print(f"{temp_fahrenheit} = {temp_celsius:.1f}")
```

```python
l = 16.4 * u.imperial.inch
print(f"{l} = {l.to(u.m):.3f}")
```

```python
E = 1 * (u.L * const.atm)
print(f"1 * (u.L * const.atm) = {E} = {E.to(u.J):.3f}")
```
