#### 3. 

Express the following temperatures in degrees Celsius.

**(a)** 9001°F

**(b)** 98.6°F (the normal body temperature of human beings)

**(c)** 20°F above the boiling point of water at 1 atm pressure

**(d)** -40°F

```python
a_fahrenheit = 9_001 * (u.imperial.Fahrenheit)
a_celsius = a_fahrenheit.to(u.deg_C, equivalencies=u.temperature())
print(f"(a) {a_fahrenheit:,.0f} = {a_celsius:,.0f}")

b_fahrenheit = 98.6 * (u.imperial.Fahrenheit)
b_celsius = b_fahrenheit.to(u.deg_C, equivalencies=u.temperature())
print(
    f"(b) {b_fahrenheit:,.1f} = {b_celsius:,.1f}, (the normal body temperature of human beings)"
)

# boiling point of water in fharenheit
bpw_f = (100 * u.deg_C).to(u.imperial.Fahrenheit, equivalencies=u.temperature())
delta_t = (20 * u.imperial.Fahrenheit) + bpw_f
c = delta_t.to(u.deg_C, equivalencies=u.temperature())
print(f"(c) 20°F above the boiling point of water at 1 atm pressure = {c:.0f}")

d_fahrenheit = -40 * (u.imperial.Fahrenheit)
d_celsius = d_fahrenheit.to(u.deg_C, equivalencies=u.temperature())
print(f"(b) {d_fahrenheit:,.0f} = {d_celsius:,.0f}")
```

page 1176

Express the following in SI units, either
base or derived.

**(a)** 55.0 miles per hour (1 mile = 1609.344 m)

**(b)** 1.15 g cm^-3

**(c)** 1.6 × 10^-19 C Å

**(d)** 0.15 mol L^-1

**(e)** 5.7 × 10^3 L atm day^-1

```python
a = 55.0 * (u.imperial.mi / u.hr)
a_si = a.to(u.m / u.s)
print(f"(a) {a:.1f} = {a_si:.1f}")

b = 1.5 * (u.g / u.cm**3)
b_si = b.to(u.kg / u.m**3)
print(f"(b) {b:.1f} = {b_si:.1e}")

c = 1.6e-19 * (u.C / u.AA)
c_si = c.to(u.C / u.m)
print(f"(c) {c:.1e} = {c_si:,.1e}")

d = 0.15 * (u.mol / u.L)
d_si = d.to(u.mol / u.m**3)
print(f"(d) {d:.2f} = {d_si:,.1e}")

e = 5.7e3 * (u.L * const.atm / u.day)
e_si = e.decompose()
print(f"(e) {e_si:,.1e} = {e_si.to(u.W):.1e}")
```
