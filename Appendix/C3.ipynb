#### 21

Use the graphical method and a calculator to solve the equation
$$log\; ln x = -x$$
for x. Give x to four significant figures.

```python
# --- 1. Numerical Solution with SymPy ---
x_sym = sp.Symbol("x", real=True, positive=True)

# Equation: ln(x) = 10^(-x)  =>  x = exp(10^(-x))
# Using exact SymPy integer sp.S(10) avoids precision mismatches
eq = x_sym - sp.exp(sp.S(10) ** (-x_sym))

# Initial guess x = 1.3
sol = sp.nsolve(eq, x_sym, 1.3)
sol_val = float(sol)

display(Math(rf"x = \mathbf{{{sol_val:.4g}}}"))


# --- 2. Graphical Method ---
x_vals = np.linspace(1.01, 2.5, 500)
y1 = np.log10(np.log(x_vals))
y2 = -x_vals

plt.figure(figsize=(7, 4.5))
plt.plot(x_vals, y1, label=r"$y = \log_{10}(\ln x)$", color="tab:blue", lw=2)
plt.plot(x_vals, y2, label=r"$y = -x$", color="tab:orange", lw=2)
plt.plot(
    sol_val,
    -sol_val,
    "ro",
    markersize=7,
    label=f"Intersection ({sol_val:.4f}, {-sol_val:.4f})",
)

plt.axhline(0, color="gray", linewidth=0.8, linestyle="--")
plt.title(r"Graphical Solution to $\log_{10}(\ln x) = -x$")
plt.xlabel("$x$")
plt.ylabel("$y$")
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
```

Use a calculator to find a number that is equal to the reciprocal
of its own natural logarithm. Report the answer to
four significant figures.

```python
x_sym = sp.Symbol("x", real=True, positive=True)

# x = 1 / ln x
eq = x_sym - (1 / sp.log(x_sym))

sol = sp.nsolve(eq, x_sym, 1.3)
sol_val = float(sol)

# Formatted to 4 significant figures as requested in Problem 21
display(Math(rf"x = \mathbf{{{sol_val:.4g}}}"))
```
