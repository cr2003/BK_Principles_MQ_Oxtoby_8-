#### Velocity, Momentum, and Acceleration

page 1166

Imagine standing at x = 2 ft of a Cartesian coordinate system and
rolling a ball towards your friend, who is standing at = 5 17.4 ft (a positive
coordinate, so your friend is in the +x direction). You could easily calculate the
speed you’d have to roll the ball in order for it to reach your friend in 3.5 seconds
(if it did not slow down en route).

```python
x_i = 2 * u.imperial.ft
x_f = 17.4 * u.imperial.ft
dt = 3.5 * u.s

speed = (x_f - x_i) / dt
print(f"speed = {speed:.1f} = {speed.to(u.imperial.mi / u.hr):.0f}")
```

Now your friend is standing at x = -13.4 ft. Let's use vectors for the calculation.

```python
pos_i = vp.vec(2, 0, 0)  # ft
pos_f = vp.vec(-13.4, 0, 0)  # ft
delta_t = 3.5  # s

# v = delta_pos / delta_t
v = (pos_f - pos_i) / delta_t  # ft/s
print(f"Velocity vector v = {v} ft/s")

speed = v.mag
print(f"Speed = {speed} ft/s")
```

#### 14

The fastball of a famous pitcher in the National League
has been clocked in excess of 95 mph. Calculate the work
done by the pitcher in accelerating the ball to that speed.
The mass of a baseball is 0.145 kg.


```python
# --- Given data ---
v = (95 * (u.imperial.mi / u.hr)).to(u.m / u.s)
m = 0.145 * u.kg

# Work-Energy Theorem: net work done on an object equals its change in kinetic energy
# W = delta KE = (1/2) m v_f^2 - (1/2) m v_i^2
# as baseball starts from rest (v_i = 0)
W = (1 / 2) * m * v**2
print(f"W = {W.to(u.J):.0f}")
```

#### 15 

A tennis ball weighs approximately 2 ounces on a postage
scale. A student practices his serve against the wall of the
chemistry building, and the ball achieves the speed of 98
miles per hour. Calculate the kinetic energy of the ball
after the serve.

```python
# --- Given data ---
m = (2 * (u.imperial.ounce)).to(u.kg)
v = (98 * (u.imperial.mi / u.hr)).to(u.m / u.s)

KE = (1 / 2) * m * v**2
KE.to(u.J)
print(f"KE = {KE.to(u.J):.0f}")
```
