#### 1

Express the following in scientific notation.

**(a)** 0.0000582

**(b)** 402

**(c)** 7.93

**(d)** -6593.00

**(e)** 0.002530

**(f)** 1.47


```python
a = 0.0000582
print(f"(a) {a:.7f} = {a:.2e}")

b = 402
print(f"(b) {b} = {b:.2e}")

c = 7.93
print(f"(c) {c} = {c:.2e}")

d = -6593.00
print(f"(d) {d} = {d:.4e}")

e = 0.002530
print(f"(e) {e} = {e:.3e}")

f = 1.47
print(f"(f) {f} = {f:.2e}")
```

#### 2. 

Express the following in scientific notation.

**(a)** 4579

**(b)** -0.05020

**(c)** 2134.560

**(d)** 3.825

**(e)** 0.0000450

**(f)** 9.814

```python
a = 4579
print(f"(a) {a} = {a:.3e}")

b = -0.05020
print(f"(b) {b} = {b:.3e}")

c = 2134.560
print(f"(c:) {c:.3f} = {c:.6e}")

d = 3.825
print(f"(d) {d} = {d:.3e}")

e = 0.0000450
print(f"(d) {e:.7f} = {e:.2e}")

f = 9.814
print(f"(f) {f} = {f:.3e}")
```

#### 3. 

Convert the following from scientific notation to decimal form.

**(a)** $5.37 × 10^{-4}$

**(b)** $9.390 × 10^{6}$

**(c)** $-2.47 × 10^{-3}$

**(d)** $6.020 × 10^{-3}$

**(e)** $2 × 10^4$

```python
a = 5.37e-4
print(f"(a) {a:.2e} = {a:.6f}")

b = 9.390e6
print(f"(b) {b:.3e} = {b:,.0f}")

c = -2.47e-3
print(f"(c) {c:.2e} = {c:.5f}")

d = 6.020e-3
print(f"(d) {d:.3e} = {d:.6f}")

e = 2e4
print(f"(d) {e:.0e} = {e:.0f}")
```

#### 4. 

Convert the following from scientific notation to decimal form.

**(a)** $3.333 × 10^{-3}$

**(b)** $-1.20 × 10^{7}$

**(c)** $2.79 × 10^{-5}$

**(d)** $3 × 10^{1}$

**(e)** $6.700 × 10^{-2}

```python
a = 3.333e-3
print(f"(a) {a:.3e} = {a:.6f}")

b = -1.20e7
print(f"(b) {b:.3e} = {b:,.0f}")

c = 2.79e-5
print(f"(c) {c:.2e} = {c:.7f}")

d = 3e1
print(f"(d) {d:.0e} = {d:.0f}")

e = 6.700e-2
print(f"(d) {e:.3e} = {e:.5f}")
```
