Calculate the following integrals.

**(a)** $\int_2^4 (3x - 1) dx$

```python
x = sp.Symbol("x")
expr = 3 * x - 1

# Symbolic definite integral
integral_expr = sp.Integral(expr, (x, 2, 4))
result = integral_expr.doit()

# Formatted LaTeX display
display(Math(rf"{sp.latex(sp.Eq(integral_expr, result))}"))
```

**(b)** $\int_0^5 x^6 dx$

```python
x = sp.Symbol("x")
expr = x**6

# Symbolic definite integral
integral_expr = sp.Integral(expr, (x, 0, 5))
result = integral_expr.doit()

# Formatted LaTeX display
display(Math(rf"{sp.latex(sp.Eq(integral_expr, result))}"))
```

**(c)** $\int_1^4 e^{-2x} dx$

```python
x = sp.Symbol("x")
expr = sp.exp(-2 * x)

# Symbolic definite integral
integral_expr = sp.Integral(expr, (x, 1, 4))
result_exact = integral_expr.doit()
result_num = result_exact.evalf(4)

# Display exact result
display(Math(rf"{sp.latex(sp.Eq(integral_expr, result_exact))}"))

# Display numerical approximation to 4 significant figures
display(Math(rf"\int_{{1}}^{{4}} e^{{-2x}} \, dx \approx \mathbf{{{result_num}}}"))
```
