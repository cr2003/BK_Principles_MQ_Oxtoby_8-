### C.2 Solution of Algebraic Equations

page 1179

$$5x + 9 = 0$$


```python
x = sp.symbols("x", real=True)

eq = sp.Eq(5 * x + 9, 0)

sol = sp.solve(eq, x)

display(Math(rf"{sp.latex(eq)}"))
display(Math(rf"x = {sp.latex(sol)}"))
```

```python
a, b, c, x = sp.symbols("a, b, c, x", real=True)

eq = sp.Eq(a * x**2 + b * x + c, 0)
sol = sp.solve(eq, x)

display(Math(rf"{sp.latex(eq)}"))
display(Math(rf"x = {sp.latex(sol)}"))
```

```python
a, b, c, x = sp.symbols("a, b, c, x", real=True)

eq = sp.Eq(x**2 - 3 * x - 7, 0)
sol = sp.solve(eq, x)

display(Math(rf"{sp.latex(eq)}"))
display(Math(rf"x = {sp.latex(sol)}"))

# Convert to real float numbers
sol_floats = [float(s) for s in sol]
print("Numerical values:", sol_floats)

# Numerical evaluation rounded to desired decimal places (e.g., 4 digits)
sol_num = [s.evalf(4) for s in sol]
display(Math(rf"x = {sp.latex(sol_num)}"))
```

#### 9

page 1194

Solve the following quadratic equations for x.

**(a)** 4x^2 + 7x -5 = 0

```python
x = sp.symbols("x", real=True)
eq_a = sp.Eq(4 * x**2 + 7 * x - 5, 0)
sol = sp.solve(eq_a, x)
display(Math(rf"(a) \;{sp.latex(eq_a)}\implies x = {sp.latex(sol)}"))

# Evaluate to floating-point real numbers
sol_numeric = [sp.N(s, 4) for s in sol]

display(Math(rf"(a) \;{sp.latex(eq_a)}\implies x = {sp.latex(sol_numeric)}"))
```

**(b)** 2x^2 = -3 - 6x

```python
x = sp.symbols("x", real=True)
eq_b = sp.Eq(2 * x**2, -3 - 6 * x)
sol = sp.solve(eq_b, x)

# Evaluate to floating-point real numbers
sol_numeric = [sp.N(s, 4) for s in sol]

display(Math(rf"(b) \;{sp.latex(eq_b)}\implies x = {sp.latex(sol_numeric)}"))
```

**(c)** $2x +\frac{3}{x} = 6$

```python
x = sp.symbols("x", real=True)
eq_c = sp.Eq(2 * x + 3 / x, 6)
sol = sp.solve(eq_c, x)

# Evaluate to floating-point real numbers
sol_numeric = [sp.N(s, 4) for s in sol]

display(Math(rf"(c) \;{sp.latex(eq_c)}\implies x = {sp.latex(sol_numeric)}"))
```
