## Formulae
### Equations of Motion with Constant Acceleration

$$v_x = v_{0x} + a_x t$$

$$x = x_0 + v_{0x} t + \frac{1}{2}a_x t^2$$

$$v_x^2 = v_{0x}^2 + 2 a_x (x - x_0)$$

$$v_{avg} = \frac{\Delta x}{\Delta t} \implies \frac{1}{2}(v_0 + v) = \frac{x-x_0}{t-t_0}$$
